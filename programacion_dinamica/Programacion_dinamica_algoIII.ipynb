# Punto 9: KingArmy

```python
#Ejemplo
n = 10
memoria = [-1 for _ in range(n+1)]
```

```python
def kingArmy(di):

    #Si el valor solicitado ya está cargado en mi estructura de memoizacion, lo retorno
    if memoria[di] != -1:
        return memoria[di]

    #Si estoy en casos base 0 y 1, debo guardarlos en mi estructura de memoizacion para que siempre caiga en la anterir cláusula
    #Esto ocurre solo una vez
    if di == 0 or di == 1:
        memoria[di] = 1
        return 1

    #Caso recursivo: busco las cantidades de los días anteriores y las sumo
    res = kingArmy(di-1) + kingArmy(di-2)
    #Guardo en la memoria el resultado
    memoria[di] = res
    #Devuelvo
    return res
```

```python
res = kingArmy(n)
print(res)
```

# Punto 10: Vacations

```python
#Ejemplo
n = 4
memo = {(dia, act): -1 for dia in range(n+1) for act in ['DESC', 'GYM', 'COMP']}
dias_gym_disponible = [2, 3]
dias_comp_disponible = [1,2]
diasGym = [False for _ in range(n)]
diasComp = [False for _ in range(n)]

for dia in dias_gym_disponible:
    diasGym[dia-1] = True

for dia in dias_comp_disponible:
    diasComp[dia-1] = True
```

```python
def puedeGym(dia, ultAct):
    if diasGym[dia] == True:
        return ultAct != 'GYM'
    else:
        return False

def puedeComp(dia, ultAct):
    if diasComp[dia] == True:
        return ultAct != 'COMP'
    else:
        return False


def vacations(dia, ultAct):

    if memo[(dia, ultAct)] != -1:
        return memo[(dia, ultAct)]

    if dia == n:
        memo[(dia, ultAct)] = 0
        return 0

    sePuedeGym = puedeGym(dia, ultAct)
    sePuedeComp = puedeComp(dia, ultAct)

    if sePuedeGym and sePuedeComp:
        res = min(1 + vacations(dia + 1, 'DESC'), vacations(dia + 1, 'GYM'), vacations( dia + 1, 'COMP'))
        memo[(dia, ultAct)] = res
        return res

    elif sePuedeGym:
        res = min(1 + vacations(dia + 1, 'DESC'), vacations(dia + 1, 'GYM'))
        memo[(dia, ultAct)] = res
        return res

    elif sePuedeComp:
        res = min(1 + vacations(dia + 1, 'DESC'), vacations(dia + 1, 'COMP'))
        memo[(dia, ultAct)] = res
        return res

    else:
        res = 1 + vacations(dia + 1, 'DESC')
        memo[(dia, ultAct)] = res
        return res
```

```python
res = vacations(0, 'DESC')
print(res)
```

Quiero reconstruir la solución para poder indicar qué se debe hacer cada día.

```python
#Yo ya conozco cuánto da res. Luego, voy consultando las soluciones mínimas para los siguientes días

actividades = []
dia = 0
aux = res
while dia < n:
    if aux == 1 + memo[(dia + 1, 'DESC')]:
        actividades.append('DESC')
        aux -= 1
    elif aux == memo[(dia + 1, 'GYM')]:
        actividades.append('GYM')
    else:
        actividades.append('COMP')

    dia += 1

print(actividades)
```

Probemos la correctitud del algoritmo por inducción en $n$.
El predicado es:

$P(n) = $ "_vacations(0,ultAct)_ computa la cantidad mínima de días que Tomas puede descansar en lugar de hacer una actividad en sus $n$ días de vacaciones después de haber hecho 'ultAct' "

Veamos primero caso base $P(0)$:

Tomas no tiene vacaciones. Luego, no tiene días de descanso.

Entonces descansa 0 días.

Como se da en la primer iteración $dia = n = 0 \Rightarrow vacations(0, ultAct) = 0 \hspace{1cm} \checkmark$


Planteemos ahora la hipótesis inductiva y nuestro objetivo.

$HI = $ "vacatios(0, ultAct) computa correctamente la cantidad mínima de días que Tomas puede descansar en lugar de hacer una actividad en sus $k$ días de vacaciones después de haber hecho 'ultAct' (es decir, vale $P(k)$)"

$Objetivo:$ ver que se cumple que se cumple $P(k+1)$

Como $0 \neq k + 1$, entramos en el caso recursivo. Notar que podemos dividir las diversas soluciones en dos:

- 1) Que en el día $0$ (para el algoritmo, el momento $(k+1) -1$, pues los días empiezan en $0$) Tomas descanse. Luego:
     - minimos dias de descanso = $1 + P(k)$ puesto que es como "renombrar" los días, y el día 1 sea con dia = 1, ... , el dia k sea dia=k (en vez de el dia 1 es el dia= 0, ..., el día k es el dia = k-1). Y en el resto de los k días buscar la mínima opción, que según $HI$ es $P(k)$

- 2) Que en el día $0$ (para el algoritmo, el momento $(k+1) -1$, pues los días empiezan en $0$) Tomas no descanse. Luego:
     - $P(k+1) = min(vacations(1, 'GYM'), vacations(1, 'COMP'))$ que por $HI$ sobre $vacations(1, 'GYM')$ y $vacations(1, 'COMP')$ se cumple que devuelve la mínima cantidad de días de descanso si en el primer día puede haber hecho cualquiera de los dos actividades en un periodo de $k$ días.
     - $P(k+1) = vacations(1, 'GYM')$ que por $HI$ sobre $vacations(1, 'GYM')$ cumple que devuelve la mínima cantidad de días de descanso si en el primer día solo puede hacer gimnasio, en un periodo de $k$ días.
     - $P(k+1) = vacations(1, 'COMP')$ que por $HI$ sobre $vacations(1, 'COMP')$ cumple que devuelve la mínima cantidad de días de descanso si en el primer solo podía hacer competencia, en un periodo de $k$ días.

Por lo tanto, se cumple el paso inductivo.


Finalmente, por principio de inducción, el algoritmo es correcto $\forall n \in \mathbb{N}$

$\hspace{25cm} \huge\boxtimes$



En cuanto a la complejidad temporal del algoritmo:

$$T(n) = O( \#estados * costo \ por \ estado)$$

Donde se tienen $N$ días con $3$ posibles propuestas por día. Luego:

Cantidad de estados = $O(3N) = O(N)$

Por otro lado, en cada instancia se hacen comparaciones, sumas y/o evaluaciones $O(1)$

Por lo tanto: $T(n) = O(N)$

Adicionalmente, si contamos el trabajo pre y post ejecución de la función, se tienen instancias adicionales de:

- Etapa previa: se genera una matriz de tamaño 3N ($O(3N)$ complejidad), y luego dos listas de tamaño $N$ ($O(N)$). Es decir, se tiene una complejidad temporal inicial de $O(3N) + O(N) = O(N)$
- Etapa posterior: para reconstruir la solución, se hace un recorrido de la matriz para evaluar cómo fue la elección de actividades, observando cuál se eligió en cada día. Luego, esto convella $O(N)$. Por lo tanto, se tiene una complejidad temporal final de $O(N)$

Es entonces que la complejidad temporal total queda: $T(n) = O(N) + O(N) + O(N) = O(N)$

#Punto 11: SumaDinamica

Veamos que son definiciones equivalentes:



$
\text{ss}(\{c_1, \ldots, c_n\}, k) =
\begin{cases}
k = 0 & \text{si } n = 0 \\
\text{ss}(\{c_1, \ldots, c_{n-1}\}, k) \lor \text{ss}(\{c_1, \ldots, c_{n-1}\}, k - c_n) & \text{si } n > 0
\end{cases}
$

y



$
ss'_C(i, j) =
\begin{cases}
j = 0 & \text{si } i = 0 \\
ss'_C(i - 1, j) & \text{si } i \neq 0 \land C[i] > j \\
ss'_C(i - 1, j) \lor ss'_C(i - 1, j - C[i]) & \text{si no}
\end{cases}
$






Si $n = 0$ luego eso significa que $0 \leq i \leq 0 \Rightarrow i = 0$, y si tomamos $j=k$ luego $ss(\{\}, k) = (k == 0) = (j == 0) = ss'_C(0, j)$

Ídem si $i = 0 \Rightarrow n = 0$, y dado que $j = k$ entonces $ss'_C(0, j) = (j == 0) = (k == 0) = ss( \{ \}, k) $

Ahora bien, dado el conjunto $C = \{c_1, ..., c_n\}, n\geq 1,$ si hay solución (llamemosla $S$), puede ocurrir que, si tomamos $j = k$, e $i = n$:

- 1) $c_n \notin S$, y por lo tanto $ss(\{c_1,...,c_n\}, k) = ss(\{c_1,...,c_{n-1}\}, k)$ (puesto que sabeos que $ss(\{c_1,...,c_{n-1}\}, k - c_n)$ es falso). Esto, por otro lado, se puede deber por dos motivos:
    - O bien $C[n] >k \Rightarrow$ entramos al caso $i \neq 0 \land C[i] > j \Rightarrow ss'_C(i,j) = ss'_C(i-1, j) \underset{i = n}{=} ss'_C(n-1, j) \ \checkmark$
    - O bien $C[n] < k \Rightarrow $ entramos al caso $i \neq 0 \land C[i] \leq j \Rightarrow ss'_C(i,j) = ss'_C(i-1,j) \underset{i = n}{=} ss'_C(n-1, j) \ \checkmark$

    Notar que sabemos que el caso $ss'_C = (n-1, j - c_n)$ es falso, y por eso no lo consideramos.

- 2) $c_n \in S$, y por lo tanto $ss(\{c_1,...,c_n\}, k) = ss(\{c_1,...,c_{n-1} \}, k - c_n)$. Además, en la otra definición, implica que entramos al caso $i \neq 0 \land C[i] < k$ por lo que entonces $ss'_C(i,j) = ss'_C(i-1, j - C[i]) \underset{i = n}{=} ss'_C(n-1, j - c_n) \ \checkmark$
  
    Notar que sabemos que el caso $ss'_C = (n-1, j)$ es falso, y por eso no lo consideramos.

Y esto es aplicable a cualquier subconjunto de $C$ que se obtenga eliminando los últimos elementos. Es decir, si luego tomo

$C \setminus$ \{$c_n$\} = \{ $c_1, ..., c_{n-1}$ }, puedo aplicar la misma lógica, y voy construyendo/obteniendo la misma solución $S$ (o, realmente, cuando $i = 0$ (caso función $ss'_C$) o bien observe $C \setminus \{c_1,...,c_n\} = \{ \}$ (caso función $ss$), alcanzaré el mismo valor de verdad $(j = k) == 0$.


La diferencia entre las dos funciones radica en la poda por _factibilidad_ (típica de implementación por backtracking, donde se descartan combinaciones antes de siquiera procesarlas) que realiza la función $ss'_C$, puesto que evalúa si para un $1\leq i \leq n$ ocurre que $C[i] > j$, entonces es ilógico pensar que, si existe $S$ subconjunto solución,se pueda dar que $C[i] = c_i \in S$ (nótese que $c_i \in \mathbb{N} \quad \forall \ 1\leq i \leq n \ $).

Sin embargo, en el peor caso, yo puedo tener un $j$ muy grande, de forma que $\forall \ 1\leq i \leq n \ $ entre al caso recursivo $ss'_C(i - 1, j) \lor ss'_C(i - 1, j - C[i])$

Es decir, en el peor caso por hago en cada instancia $i$ dos llamados recursivos.

Luego, en total serían $\underset{n \ veces}{O(2*2*...*2)} = O(2^n)$ llamados recursivos en el peor caso

Por otro lado, en esta segunda definición, el conjunto $C$ es usado como variable global, es decir, no es necesario llamarla en la función y, por otro lado, no se ve modificada.

Luego, si tomamos $n = len(C)$ y $k$ la suma a obtener, para cada $1 \leq i \leq n$, se puede dar que la función $ss'_C$ sea llamada con $1 \leq j \leq k$.

Es decir, hay $n*k$ formas de llamar a la función, o dicho de otro modo, combinaciones de variables (_estados_) que pueden darse.

Es decir, existen $O(nk)$ entradas para la función $ss'_C$

Ahora bien, el algoritmo visto presentaba una cantidad $O(2^n)$ llamadas en el peor caso.

Entonces, si $kn << 2^n \iff k << \frac{2^n}{n}$ entonces necesariamente deberán haber entradas $(i,j)$ que se repitan utilizando ese algoritmo. Es decir, si $k << \frac{2^n}{n}$ inevitablemente habrá llamadas recursivas que resuelvan el mismo problema.


Hagamos un ejemplo:

Sea $k = 1$, $\ C = \underset{100 \ veces}{\{1, 1, 1, ..., 1\}} \ \ $ ($1 << \frac{2^{100}}{100} $)

Luego, $\forall \ 1 \leq i \leq 100$ se da que $i \neq 0 \land C[i] <= 1$

Y luego, cuando $i = 100$, se tendrá que:

$ss'_C(100, 1) = ss'_C(100-1, 1) \lor ss'_C (100-1, 0)$

Luego

$ss'_C(100-1, 1) = ss'_C(100 -1 -1, 1) \lor ss'_C(100 -1 -1, 0)$

y

$ss'_C(100 -1, 0) = ss'_C(100 -1 -1, 0)$

y

$ss'_C(100 -1-1, 1) = ss'_C (100 -1 -1 -1, 1) \lor ss'_C (100 -1 -1 -1, 0)$

y

$ss'_C(100 -1 -1, 0) = ss'_C(100 -1 -1 -1, 0)$

$\vdots$

Y si se continúa desarrollando, se obserbarán muchas repeticiones más.

Consideremos luego la estructura de memoización (i.e., el diccionario) $M$ implementada como una matriz de $(n+1)×(k+1)$ tal que $M[i,j]$ o bien tiene un valor indefinido $⊥$ o bien tiene el valor $ss'_C(i,j)$, para todo $0 ≤ i ≤ n \ $ y $\ 0 ≤ j ≤ k$.

Y sea el siguiente algoritmo top-down de $ss'_C(i, j)$

1) Inicializar $M[i,j] = ⊥$ para todo $0 ≤ i ≤ n \ $ y $ \ 0 ≤ j ≤ k$.
2) $subset\_sum(C,i,j):$ // implementa $ss({c1,...,ci},j) = ss'_C(i,j)$ usando memoización
3) Si $j < 0$, retornar falso
4) Si $i = 0$, retornar $(j == 0)$
5) Si $M[i,j] = ⊥$:
6) Poner $M[i,j] = subset\_sum(C,i −1,j) ∨ subset\_sum(C,i−1,j −C[i])$
7) Retornar $M[i,j]$

Veamos si mantiene un estado válido para $M$ y computa $M[i,j] = ss'_C(i,j)$ cuando se invoca $ss'_C(i, j)$

Primero es importante notar que como $0 \leq i \leq n \ $ y $ \ 0 \leq j \leq k$, entonces en total hay $(n+1)*(k+1)$ estados, y este es justamente el tamaño de $M$. Es decir, hay suficiente espacio para almacenar cada uno.

Por otra parte, nótese que en la matriz $M$, si se incia la función con $i = n$ y $j = k$, entonces $M$ no guarda estados con $j,i < 0$ y tampoco $i >n$ ni $j >k$.

Es decir, cada celda de nuestra matriz corresponderá entonces al resultado de [$j == 0$ si $i = 0$] o [$subset\_sum(C,i −1,j) ∨ subset\_sum(C,i−1,j −C[i])$] para entradas válidas de $i,j$.

Por otra parte, cualquiera sea la entrada válida $(i,j)$, $M[i][j]$ almacena $ss'_C(i,j)$ puesto que si $i = 0$, almacenará $j == 0$, que es el valor de verdad que devuelve $ss'_C(0, j)$ cualquiera sea el $1 \leq j \leq k$.

Y por otra parte, si $i > 0$, entonces $M[i][j]$ almacenará el resultado de $subset\_sum(C,i −1,j) ∨ subset\_sum(C,i−1,j −C[i])$ que es equivalente al caso $n > 0$ de la función $ss(C, j)$ con $C = \{ c_1,...,c_i \}$ que, como mencionamos, engloba (equivale) a los dos casos recursivos de $ss'_C$ para la entrada $(i,j)$.

Luego, como $subset\_sum$ es una implementación válida de $ss'_C$ (o de $ss$), entonces resuelve el problema de suma de subconjuntos.



Por otro lado, al evaluar complejidad, nótese que para cada par $(i, j)$, la llamada $ss'_C(i, j)$ se resuelve una _única_ vez, puesto que luego es almacenado (guardado) el resultado en la estructura de memoización.
Es entonces que, a lo sumo, ocurrirán todos los $O(n*k)$ estados, y el costo de cada uno es $O(1)$ (correspondiente al guardado de la información en la matriz y comparación de valores booleanos).

Luego, la complejidad temporal es $O(nk)$

Y, por lo tanto, si $k << 2^{n}$, es más eficiente la implementación top-down y, en caso contrario, la implementación backtracking.

# Punto 12: Optipago


Tenemos un multiconjunto $B$ de valores de billetes y queremos comprar un producto de costo $c$ de una máquina que no da vuelto. Para poder adquirir el producto debemos cubrir su costo usando un subconjunto de nuestros billetes. El objetivo es pagar con el mínimo exceso posible a fin de minimizar nuestra pérdida. Más aún, queremos gastar el menor tiempo posible poniendo billetes en la máquina. Por lo tanto, entre las opciones de mínimo exceso posible, queremos una con la menor cantidad de billetes. Por ejemplo, si $c = 14$ y $B = \{2,3,5,10,20,20\}$, la solución es pagar 15, con exceso 1, insertando sólo dos billetes: uno de 10 y otro de 5.

Consideramos luego la siguiente estrategia por backtracking para el problema, donde $B = {b_1,...,b_n}$.

Tenemos dos posibilidades: o agregamos el billete $b_n$, gastando un billete y quedando por pagar $c - b_n$, o no agregamos el billete $b_n$, gastando $0$ billetes y quedando por pagar $c$.

Entonces, la función queda:

$
\text{cc}(\{b_1, \ldots, b_n\}, c) =
\begin{cases}
(0, 0) & \text{si } c \leq 0\\ (\infty , \infty ) & \text{si } n = 0 \ \land \ c> 0\\
cc(\{b_1,...,b_{n-1}\}, c) & \text{si } m < m' + b_n \  \lor \ (m = m' + b_n \ \land \ q \leq q' + 1) \\ (m' + b_n, 1 + q') & sino
\end{cases}
$

donde $m = cc(\{b_1, \ldots, b_{n-1}\}, c)[0]$

$\hspace{1.3cm}q = cc(\{b_1, \ldots, b_{n-1}\}, c)[1]$

$\hspace{1.3cm} m' = cc(\{b_1, \ldots, b_{n-1}\}, c - b_n)[0]$

$\hspace{1.3cm}q' = cc(\{b_1, \ldots, b_{n-1}\}, c - b_n)[1]$

```python
#Implementación de cc({b1,...,bi},j)

def cc(B, j):
    i = len(B)

    if j <= 0:
        return (0,0)

    if i == 0 and j>0:
        return (float('inf'), float('inf'))

    (m, q) = cc(B[:i-1], j)
    (m_prima, q_prima) = cc(B[:i-1], j - B[i-1])

    if (m < m_prima + B[i-1]) or ((m == m_prima + B[i-1]) and (q <= q_prima + 1)):
        return (m,q)

    else:
        return (m_prima + B[i-1], q_prima + 1)
```

```python
#Ejemplo
c = 14
B = [2,3,5,10,20,20]

monto_minimo_a_pagar, cantidad_de_billetes = cc(B, c)
print(f'El monto mínimo a pagar es {monto_minimo_a_pagar} y se obtiene usando como mínimo {cantidad_de_billetes} billetes')
```

Ahora bien, nótese que en el peor caso, en cada instancia se ejecutará el paso recursivo, es decir, se realizarán dos llamadas recursivas por cada invocación de la función.

Luego, eso quiere decir que habrán $O(2^N)$ llamadas recursivas en total.

Dado que en cada una, solo se realizan operaciones $O(1)$ (sumas, comparaciones), eso genera que la complejidad temporal en el peor caso sea:

$T(n) = O(2^N)*O(1) = O(2^N)$

Sea:

$
cc'_B(i, j) =
\begin{cases}
(0, 0) & \text{si } j \leq 0\\ (\infty , \infty ) & \text{si } i = 0 \ \land \ j> 0\\
cc'_B(i-1, j) & \text{si } m < m' + b_i \  \lor \ (m = m' + b_i \ \land \ q \leq q' + 1) \\ (m' + b_i, 1 + q') & sino
\end{cases}
$

donde $m = cc'_B(i-1, j)[0]$

$\hspace{1.3cm}q = cc'_B(i-1, j)[1]$

$\hspace{1.3cm} m' = cc'_B(i-1, j - b_i)[0]$

$\hspace{1.3cm}q' = cc'_B(i-1, j - b_i)[1]$


Notar que ésta es una reescritura de $cc$ de forma recursiva que implementa la
idea anterior dejando fijo el parámetro $B$.

Esta nueva definición de la función permite observar que hay $0(Nc)$ estados diferentes de la función, puesto que $0 \leq i \leq N \ $ y $ \ 0 \leq j \leq c$

Por ende, si $c << \frac{2^N}{N}$ (o, en definitiva, $c << 2^N$), habrá combinaciones de entradas $(i,j)$ que se repitirán.

Luego, si planteamos la estructura de memoización $M$, con $M$ una matriz de $(N+1)*(c+1)$ donde $M[i][j] = cc'_B(i,j)$ (donde, en nuestro caso, la llamada $cc'_B(N, c)$ es la que resuelve el problema original), entonces seríamos capaces de acceder en $O(1)$ a llamadas recursivas que ya han sido realizadas en pasos anteriores, y lograr así mejorar la complejidad de nuestra anterior implementación.

```python
#Desarrollo un nuevo algoritmo implementando esta idea

#Primero debo crear mi estructura de memoización
filas = len(B) + 1
columnas = c + 1
# Creo la matriz con (-1, -1) en cada posición
M = [[(-1, -1) for _ in range(columnas)] for _ in range(filas)]

#Ahora desarrollo la función
def cc_B (i, j):

    if j <= 0:
        return (0,0)

    if M[i][j] != (-1, -1):
        return M[i][j]

    if i == -1 and j>0:
        return (float('inf'), float('inf'))

    (m, q) = cc_B(i-1, j)
    (m_prima, q_prima) = cc_B(i-1, j - B[i])

    if (m < m_prima + B[i]) or ((m == m_prima + B[i]) and (q <= q_prima + 1)):
        M[i][j] = (m,q)
        return (m,q)

    else:
        M[i][j] = (m_prima + B[i], q_prima + 1)
        return (m_prima + B[i], q_prima + 1)
```

```python
#Usamos el mismo ejemplo de antes
monto_minimo_a_pagar, cantidad_de_billetes = cc_B(len(B) - 1, c)
print(f'El monto mínimo a pagar es {monto_minimo_a_pagar} y se obtiene usando como mínimo {cantidad_de_billetes} billetes')
```

Veamos la complejidad temporal de este algoritmo.

Gracias a mi estructura de memoización, no habrá más resoluciones de una misma llamada recursiva en repetidas oportunidades; sino que se consultará en la matriz $M$ el resultado de la misma. Esto convella un tiempo $O(1)$

Luego, el peor caso sería el cálculo de los $O(Nc)$.

Entonces, dado que en cada uno de estos estados, se llevan a cabo operaciones $O(1)$ (las mismas sumas y comparaciones de antes), entonces esto signfica que:

$T'(n) = O(Nc) * O(1) = O(Nc)$

Lo cual es muchísimo más eficiente que $O(2^N)$

Por otro lado, en cuanto a la complejidad espacial, tenemos el costo de la matriz $M$, lo cual será $O(Nc)$

(Nótese que el costo de $B$ es $O(N)$. Luego $O(Nc) + O(N) = O(Nc)$ )

# Punto 13: Astropaid

Astro Void se dedica a la compra de asteroides. Sea $p \in \mathbb{N}^n$ tal que $p_i$ es el precio de un asteroide el $i$-ésimo día en una secuencia de $n$ días. Astro Void quiere comprar y vender asteroides durante esos $n$ días de manera tal de obtener la mayor ganancia neta posible. Debido a las dificultades
que existen en el transporte y almacenamiento de asteroides, Astro Void puede comprar a lo sumo un asteroide cada día, puede vender a lo sumo un asteroide cada día y comienza sin asteroides. Además, el Ente Regulador Asteroidal impide que Astro Void venda un asteroide que no haya comprado. Queremos encontrar la máxima ganancia neta (m.g.n) que puede obtener Astro Void respetando las restricciones indicadas.

Notar que en una solución óptima, Astro Void debe terminar sin
asteroides.

Sea luego $c$ la cantidad de asteroides de Astro Void al finalizar el día $j$. Notar que:

 - No puede darse que $c < 0$ (es decir, está indefinido) pues eso indicaría que hubo mayor cantidad de ventas que de compra de asteroides, lo cual no se permite; o que arrancó con una deuda de asteroides (lo cual tampoco se permite). Luego, no existe máxima ganancia neta en este caso.
 - Tampoco puede ser que $c > j$, puesto que en cada día se puede comprar a lo sumo un asteroide. Y, por lo tanto, esto indicaría que o bien arrancó con al menos un asteroide (lo cual no es posible), o bien hubo compra de múltiples asteriodes en un solo día (que no es válido). Por lo tanto, estamos en un caso indefinido, no existe máxima ganancia neta.
 - En caso contrario, dado que en un día $j$ se puede o comprar un asteroide, o vender un asteroide, o simplemente no realizar operaciones; entonces la máxima ganancia neta es el máximo entre:
      
      - Finalizar el anterior día ($j-1$) con $c-1$ asteroides, y luego comprar uno en el día $j$
      - Finalizar el anterior día ($j-1$) con $c+1$ asteroides, y luego vender uno en el día $j$
      - Finalizar el anterior día ($j-1$) con $c$ asteroides, y no operar el día $j$.

Sin embargo, cuando $j = n$, no se puede haber alcanzado la máxima ganancia neta si $c>0$, puesto que si llamo $g$ a mi $mgn$ y tomo $p_{k_1}, p_{k_2},...,p_{k_c} \in \mathbb{N}$ los costos de mis asteroides comprados los días $k_1, k_2, ..., k_c$ y que luego no vendí, entonces si en cambio esos días hubiera optado por no adquirirlos, tendría que $mgn = g + p_1 + p_2 + ... + p_c > g = mgn \ \ ABS$

Eso implica que, cuando $j = n$, entonces la $mgn$ es el máximo entre:
  - Finalizar el anterior día ($n-1$) con un asteroides, y luego venderlo en el día $n$
  - Finalizar el anterior día ($n-1$) con ningún asteroide, y no operar el día $n$.

Pero el caso "Finalizar el anterior día ($j-1$) con $c-1$ asteroides, y luego comprar uno en el día $j$" con $2 \leq j \leq n-1 \ $ es un caso fundamental a observar en los días intermedios.

Por otro lado, en el primer día, al tener la restricción de comenzar sin poseer ningún asteroide (y al no poder vender asteroides que no tengo), entonces solo es posible comprar uno y venderlo otro día, o directamente no operar.

A partir de estos análisis, podemos desarrollar la siguiente función recursiva:

$
mgn_P(i, c) =
\begin{cases}
 0 & \text{si } i = 0 \ \land \ c= 0\\ - \infty & \text{si } c < 0 \ \lor \ c > i \ \lor (i = 0 \ \land \ c>0)\\ max \{ mgn_P(i-1, c+1) + p_i, mgn_P(i-1, c-1) - p_i, mgn_P(i-1, c)\} & \text{caso contrario}
\end{cases}
$

donde $P = \{ p_1, p_2,...,p_n\}$ con $p_i$ el precio de compra/venta de un asteroide el día $i$

Y luego, $mgn_P(n, 0)$ es la respuesta de nuestro dilema.

```python
#Implementemos la función.
#Realizaré un algoritmo de programación dinámica top-down

P = [3,2,5,6]
n = len(P)
A = [[None for _ in range(n+1)] for _ in range(n+1)]
def mgn_P(i, c):
    if c < 0 or c > i:
        return float('-inf')

    if A[i][c] is not None:
        return A[i][c]

    if i == 0 and c>0:
        A[i][c] = float('-inf')
        return float('-inf')

    if i == 0 and c == 0:
        A[i][c] = 0
        return 0

    vender = mgn_P(i-1, c+1) + P[i-1]
    comprar = mgn_P(i-1, c-1) - P[i-1]
    no_operar = mgn_P(i-1, c)

    maxima_ganancia = max(vender, comprar, no_operar)

    A[i][c] = maxima_ganancia

    return maxima_ganancia
```

```python
ganancia_maxima = mgn_P(n, 0)

print(f'La máxima ganancia neta que puede lograr Astro Void es {ganancia_maxima}')
```

Analicemos las complejidades que presenta el algoritmo.

Por un lado, tengo que en todo momento $0 \leq i \leq n \ $y $ \ -1 \leq c \leq n/2$
Esto es debido a que vamos de forma descendente desde el día $n$ hasta el momento $0$, y la cantidad de asteroides nunca baja de $-1$ (pues en cuyo caso, corta esa rama de la recursión al no ser posible alcanzar un máximo), y no será posible tener una cantidad $c > n/2$ asteroides sin que $i < n/2$, puesto que solo es posible adquirir un asteroide por día.

Luego, en el peor caso, tenemos $O(n^2)$ estados distintos.

Y dado que en cada uno se realizan operaciones $O(1)$ como sumar, restar o tomar máximo, luego se tiene que la complejidad temporal final del algoritmo es $O(n^2)$.

Por otra parte, en cuanto a la complejidad espacial, se cuenta con una matriz $A$ de dimensiones $(n+1)*(n+1)$, donde cada valor es de tamaño $O(1)$.

Luego, eso genera una complejidad espacial de $O(n^2)$

# Punto 14: Fire

Polycarp está en problemas. ¡Su casa se incendia!. Es tiempo de salvar los artículos más valiosos. Cada artículo tiene los parámetros:
- El tiempo $t$ entero que estima el sabueso que le tomara salvarlo.
- El tiempo $d$ entero que estima el sabueso a partir del cuál se quema el articulo y no sirve más (inclusive si lo está rescatando en ese momento)
- El valor $p$ del artículo para el sabueso.

El sabueso quiere maximizar la suma de los valores de los artículos que puede salvar, considerando que:
- Salva un árticulo luego de otro, uno a la vez.
- Si un artículo $A$ le toma $t_a$ segundos y luego salva el $B$, entonces le habrá tomado $t_A +t_B$ segundos en total.

Planteemos primero la función.

Supongamos que tenemos un arreglo $articulos = \{art_1, art_2,...,art_N \}$ de tamaño $N$ donde cada $art$ que lo compone es una tripla $(t, d, p)$.

Luego, yo voy a querer obtener el subconjunto ordenado $\{ art_{i_1}, art_{i_2}, ..., art_{i_k}\}$ de $articulos$ que cumpla que:
- $1)$ $\sum_{j=1}^r art_{i_j}.t < r.d \quad \forall \ 1\leq r \leq k  $

- $2)$ Para todo subconjunto $ \{ art_{y_1}, ..., art_{y_s}\}$ de $articulos$ que cumpla $1)$, se tiene que $\sum_{j = 1}^s art_{y_j}.p \leq \sum_{t=1}^k art_{i_t}.p$  

Ahora bien, en principio, $articulos$ no tiene por qué estar ordenado en función de $d$, que es el parámetro importante a tener en cuenta al momento de definir si es posible o no salvar los artículos en función de cuánto tiempo hemos ya gastado salvando otros.

Es por eso que, en principio, nos va a convenir ordenar los artículos en función del tiempo máximo en el que pueden ser salvados, en forma creciente.

Esto lo podemos lograr en $O(NlogN)$ utilizando la función $mergeSort$.

```python
# Primero usaremos merge_sort para ordenar la lista de artículos
# El algoritmo lo obtuve de asimov.cloud

def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    # 1️⃣ División
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    # 2️⃣ Fusión
    return merge(left, right)


def merge(left, right):
    merged = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1
    # Añadir los restos (solo uno de los dos bucles se ejecutará)
    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged
```

```python
#Vamos con un ejemplo
articulos = [(3,7,4), (2,6,5) ,(3,7,6)]
A = merge_sort(articulos)
print(A)
```

Sea entonces $A = merge\_sort(articulos)$

Luego, teniendo nuestros artículos en orden, pensemos en la función que resuelve el problema de hallar el valor máximo que se puede obtener en la suma descripta en $2)$.

```python
# Esta función devuelve el valor total máximo que se puede salvar.
N = len(articulos)
D = max(d for (t,d,p) in A)
memo = [[-1 for _ in range(D+1)] for _ in range (N+1)]
def fire(i, tAcum):
    if i == N:
        return 0

    if tAcum + A[i][0] >= A[i][1] :
        return float('-inf')

    if memo[i][tAcum] != -1:
        return memo[i][tAcum]

    no_salvar_art = fire(i+1, tAcum)

    #Solo puedo salvar si tAcum no supera A[i][1], que es el tiempo que tiene el art antes de volverse obsoleto
    salvar_art = float('inf')
    if tAcum + A[i][0] < A[i][1]:
        salvar_art = fire(i+1, tAcum + A[i][0]) + A[i][2]

    valorMaximo = max(no_salvar_art, salvar_art )
    memo[i][tAcum] = valorMaximo

    return valorMaximo
```

```python
#Aplicamos la función a nuestro ejemplo
valor_maximo = fire(0, 0)
print(f'El valor máximo que se puede obtener es {valor_maximo}')
```

Por último, es necesario reconstruir la solución.

Para ello, emplearé mi estructura de memoización para ir obteniendo aquellos artículos que contribuyeron al valor máximo alcanzado, y que por lo tanto significa que fueron salvados.
Dado $1 \leq i \leq N-1$, un artículo ***no*** habrá sido salvado si $A[i][tAcum] = A[i+1][tAcum]$ pues esto implicaría que la solución no consideró el tiempo que se gasta al salvar el artículo $i$ y, por lo tanto, no se lo salvó.

```python
#%% Reconsruyamos ahora la solución.
articulos_salvados = []
tAcum = 0
for i in range(0, N, 1):
    if memo[i][tAcum] != memo[i+1][tAcum]:
        tAcum += A[i][0]
        articulos_salvados.append(A[i])

print(articulos_salvados)
```

Nótese que dado que $A$ está ordenado en función del tiempo $d$ que tiene como máximo el objeto para ser salvado, entonces el orden que presenta la variable $articulos\_salvados$ describe la forma óptima de ir eligiendo/salvando los artículos, a fin de conseguir salvar el valor máximo posible.

Evaluemos ahora la complejidad temporal del algoritmo.

Antes de invocar a la función fire, en tiempo $O(NlogN)$ se ordenó la lista $articulos$ (conseguiendo la variable $A$ como resultado).

Posteriormente, se recorre la lista para obtener la duración máxima, es decir, de la lista de artículos, se seleccionó el mayor valor de duración (i.e., tiempo que tiene para no volverse obsoleto). Llamemosla $D$.

Tras esto, creamos una matriz de dimensiones $D*N$, con valores O(1) en cada celda, lo que convella $O(D*N)$.

Recién entonces, es invocada la función, la cual presenta a lo sumo $O(D*N)$ estados posibles dado que las entradas $(i, tAcum)$ posibles cumplen $0 \leq i \leq N \ $ y $ \ 0 \leq tAcum \leq D \ $ dado que no sería posible salvar ningún elemento pasado la duración máxima $D$ (ya que para cada artículo $i$, es necesario para salvarlo que el tiempo transcurrido hasta el momento no haya superado su tiempo de resistencia).

Luego, como hay $O(D*N)$ estados posibles, donde en cada uno se realizan operaciones $O(1)$ como sumas, comparaciones o selección de máximo entre dos opciones, entonces se tiene que el algoritmo convella una complejidad $O(D*N)$.

Finalmente, con el fin de reconstruir la solución, se recorre la lista de artículos para poder identificar aquellos que fueron salvados, mediante operaciones $O(1)$ como consultas a nuestra estructura de memoización o comparaciones. Esto entonces toma $O(N)$

Es así que, en cojunto:

$T(N) = O(NlogN) + O(N) + O(D*N) + O(D*N) + O(N) = O(D*N + NlogN)$

# Puntos 15: Cortes Económicos

Debemos cortar una vara de madera en varios lugares predeterminados. Sabemos que el costo de realizar un corte en una madera de longitud $ℓ$ es $ℓ$ (y luego de realizar ese corte quedarán 2 varas de longitudes que sumarán $ℓ$). Por ejemplo, si tenemos una vara de longitud 10 metros que debe ser cortada a los 2, 4 y 7 metros desde un extremo, entonces los cortes se pueden realizar, entre
otras maneras, de las siguientes formas:
- Primero cortar en la posición 2, después en la 4 y después en la 7. Esta resulta en un costo de 10 + 8 + 6 = 24 porque el primer corte se hizo en una vara de longitud 10 metros, el segundo en una de 8 metros y el último en una de 6 metros.
- Cortar primero donde dice 4, después donde dice 2, y finalmente donde dice 7, con un costo de 10 + 4 + 6 = 20, que es menor.

Queremos encontrar el mínimo costo posible de cortar una vara de longitud $ℓ$.

Notar que, dado una vara que abarca desde $i$ hasta $j$ (es decir, longitud de la vara es $j-i$) entonces el costo mínimo de cortar (si es que hay cortes por hacer entre esos límites) está formado en parte por $j-i$. Y luego, si llamamos $C = \{ c_1, c_2, ..., c_k\}$ a los cortes a realizar, entonces:

$costo\_minimo(i,j) = (j-i) + \underset{c \in C : \ i \leq c \leq j}{min} \{costo\_minimo(i,c) + costo\_minimo(c,j) \}$

Puesto que realizar un corte mínimo entre $i$ y $j$ implica luego que la posición elegida $c$ de corte genere también la forma mínima de cortar (con las posiciones pendientes) las nuevas varas obtenidas.

Gracias a esta observación, y teniendo en cuenta que cuando $i = j$, el costo del corte es $0$, entonces:

$
costo\_minimo_C(i, j) =
\begin{cases}
 0 & \text{si } \nexists c \in C : i < c < j \ \\ j-i + \underset{c \in C : i < c < j}{min} \{ costo\_minimo_C(i, c) + costo\_minimo_C(c, j)\} & \text{caso contrario}
\end{cases}
$

donde $i$, $j$ son posiciones en la vara original, y $C$ es el conjunto de posiciones $c$ de cortes a realizar.

Notar que en el caso base, si no hay un corte disponible para hacer entre los índices, no hay posibilidad de corte en medio, por lo que el costo es 0.

En caso contrario, si hay corte(s) disponible(s), entonces debo adicionar la longitud de mi vara ($j-i$) con el mínimo corte $c$ tal que se minimice la suma de costos de corte en la vara $(i,c)$ y la vara $(c,j)$.

```python
# Algoritmo dinámico Top Down
cortes = [2,4,7]
l = 10
C = [i if i in cortes else -1 for i in range(l)]
M = [[-1 for _ in range(l+1)] for _ in range(l+1)]
def costo_minimo(i,j):
    if M[i][j] != -1:
        return M[i][j]

    costos = []
    largo_vara = j - i
    for k in range(i+1, j, 1):
        coste_k = 0
        if C[k] != -1:
            coste_k = largo_vara + costo_minimo(i, k) + costo_minimo(k, j)
            costos.append(coste_k)

    minimo_corte = 0
    if costos:
        minimo_corte = min(costos)
    M[i][j] = minimo_corte
    return minimo_corte

print(costo_minimo(0, l))
```

Ahora bien, la complejidad de este algoritmo es $O(L^3)$ (donde $L$ es el tamaño de la vara original) puesto que se tiene una cantidad total de $O(L^2)$ estados, donde cada uno presenta un costo $O(L)$, proveniente del bucle.

Por otra parte, la complejidad espacial auxiliar es de $O(L^2)$, que es el costo de la matriz de memoiación.

Sin embargo, esto es optimizable si se tiene en cuenta que los cortes a realizarse están definidos por $C$.

Es decir, podemos pasar de una cantidad de estados $O(L^2)$ a una de $O(|C|^2)$, donde cada uno presente un costo de $O(|C|)$; y así conseguir una complejidad temporal final de $O(|C|^3)$

Y, por otro lado, dado que tendríamos $O(|C|^2)$ estados, entonces esa podría ser la complejidad del espacio auxiliar.

```python
#Algoritmo dinámico optimizado top-down
#Estoy teniendo en cuenta que la cadena de cortes a realizarse viene ordenada.
#Caso contrario, se puede realizar un paso previo de ordenamiento en O(|C|log|C|) que no alteraría la complejidad final.

cortes= [2,4,7]
C = [0]
l = 10
for corte in cortes:
    C.append(corte)
C.append(l)
M = [[-1 for _ in range(len(C)+1)] for _ in range(len(C)+1)]
def costo_minimo(i,j):
    if M[i][j] != -1:
        return M[i][j]

    izq = C[i]
    der = C[j]
    costos = []
    largo_vara = der - izq
    for k in range(i+1, j, 1):
        coste_k = 0
        coste_k = largo_vara + costo_minimo(i, k) + costo_minimo(k, j)
        costos.append(coste_k)

    minimo_corte = 0
    if costos:
        minimo_corte = min(costos)
    M[i][j] = minimo_corte
    return minimo_corte

print(costo_minimo(0, len(C)-1))
```

```python
#Algoritmo dinámico Bottom-up

cortes= [2,4,7]
l = 10
M = [[0 for _ in range(len(cortes)+ 2)] for _ in range(len(cortes)+2)]
def costo_minimo(l, cortes):
    C = [0]
    l = 10
    for corte in cortes:
        C.append(corte)
    C.append(l)
    for d in range(2, len(C)+1, 1):
        for i in range(0, len(C)-1-d + 1):
            j = i + d
            M[i][j] = float('inf')
            costos = []
            for k in range(i+1,j,1):
                coste_k = C[j] - C[i] + M[i][k] + M[k][j]
                costos.append(coste_k)

            if costos:
                M[i][j] = min(costos)
            else:
                M[i][j] = 0

    return M[0][len(C) - 1]

print(costo_minimo(l, cortes))
```

# Punto 16: Travesía Vital

Hay un terreno, que podemos pensarlo como una grilla de $m$ filas y $n$ columnas, con trampas y pociones. Queremos llegar de la esquina superior izquierda hasta la inferior derecha, y desde cada casilla sólo podemos movernos a la casilla de la derecha o a la de abajo. Cada casilla $i$,$j$ tiene un número entero $A_{i,j}$ que nos modificará el nivel de vida sumándonos el número $A_{i,j}$ (si es negativo, nos va a restar $|A_{i,j}|$ de vida). Queremos saber el mínimo nivel de vida con el que debemos comenzar tal que haya un camino posible de modo que en todo momento nuestro nivel de vida sea al menos 1. Por ejemplo, si tenemos la grilla:

$A =
\begin{bmatrix}
-2 & -3 & 3 \\
-5 & -10 & 1 \\
10 & 30 & -5
\end{bmatrix}
$

el mínimo nivel de vida con el que podemos comenzar es 7 porque podemos realizar el camino que va todo a la derecha y todo abajo

Nótese que para la posición $i,j$ hay dos opciones: derecha o abajo.
Luego, si $A_{0,0}$ es positivo, entonces en principio no requiero una cantidad inicial positiva es decir, en principio, $cantidad\_inicial\_necesaria = -A_{0,0} + 1$. Por otro lado, si $A_{0,0}$ es negativo, entonces en principio requiero $cantidad\_inicial\_necesaria = |A_{0,0}| + 1$

Luego, puedo elegir ir hacia abajo o hacia la derecha, si es que $1 \leq m$ y $1 \leq n$ (en caso contrario, solo habría una única opción disponible).

En ambos casos, deberé evaluar nuevamente el signo del elemento de la celda. Si es positivo, entonces no es necesario agregar cantidad alguna; en cambio, si tengo un elemento negativo, llámese $k$, entonces es necesario sumarle $|k|$ a la cantidad mínima contabilizada hasta el momento.

Finalmente, cuando no pueda ir hacia la derecha ni hacia abajo, significa que estoy en la posición $A_{m,n}$; donde nuevamente, si el elemento es positivo, no debo de realizar ninguna adición, y simplemente devolver el resultado; en caso contrario, debo sumarle $|A_{m,n}|$ al costo mínimo, para después devolverlo.

La solución(es) óptima(s) será(n) aquella(s) que devuelva(n) el mínimo costo inicial.

Convencerse de que, excepto que estemos en los límites del terreno, la mínima vida necesaria
al llegar a la posición i,j es el resultado de restar al mínimo entre la mínima vida necesaria
en i +1,j y aquella en i,j + 1, el valor Ai,j, salvo que eso fuera menor o igual que 0, en
cuyo caso sería 1.

Ahora bien, una forma de pensar la función es la siguiente:
 - Casos Base:
    
    - Si estoy en la posición $A_{m,n}$ luego si éste es positivo, por regla el mínimo de vida inicial necesaria con la que debo haber llegado es $1$; pero si es negativo, requiero incrementar la cantidad de vida con la que llego a la celda; y puesto que lo que pierdo de vida es el valor de mi celda, entonces en definitiva requiero entrar con $|A_{m,n}| + 1$ de vida.
  
  - Caso recursivo:

    - En el caso recursivo, pueden ocurrir tres cosas:

      - Estoy en una celda $A_{i,j}$ con $i < m \ $ y $ \ j < n$. Es decir, puedo a continuación bajar o ir a la derecha. Ahora bien, ¿cuál convendrá? Pues será justamente el que requiera un mínimo nivel de vida, llamemosle $v$ a ese mínimo. Una vez determinado "la necesidad futura", la necesidad actual se basa en poder entrar a la celda $A_{i,j}$ con al menos uno de vida. Notar que si $A_{i,j}$ es una poción, entonces me basta con entrar con un nivel de vida equivalente a $max \{v - A_{i,j}, 1 \}$ puesto que, si el máximo es v- A_{i,j} significa que cierta porción de la necesidad futura la cubre la poción, pero sigo teniendo un remanente que debo cubrirlo externamente; en cambio, si el máximo es 1, eso significaría que la poción satisface toda esta necesidad a posteriori, y por lo tanto el mínimo nivel de vida necesario para entrar a esta celda (y soportar el venidero veneno) es 1.

        Por otra parte, si en mi celda hay veneno ($A_{i,j} < 0$) implicaría que a $v$, el mínimo nivel de vida que requiero a futuro, debo adicionarle el valor absoluto de mi celda actual. Pues yo se que así, al entrar a la celda y consumir el veneno, me quedaría un nivel $v$ ($v \geq 1$) con lo que afrontar el resto de la jornada (y por definición de $v$, dependiendo del camino a elegir, es suficiente para mantenerse a salvo). Entonces, en definitiva, el mínimo nivel de vida necesario es $v - A_{i,j}$.

        Es notorio que, en resumen, para una celda $A_{i,j}$ donde puedo realizar ambos movimientos, entonces el mínimo de vida necesaria para entrar es, $max ${$ v - A_{i,j}, 1 \ $} con $v$ la mínima necesidad venidera.
    
    - En los casos $i=m \ $ o $ \ j = n$ lo único a tener en cuenta es que el futuro está unívocamente definido (bajar caso $i = m$, derecha caso $j = n$). Luego, $v$ representa simplemente la (única) necesidad venidera, pero la mínima vida necesaria al entrar a la celda sigue siendo $max  \{ v - A_{i,j}, 1 \} $.

Es así como obtenemos la función recursiva:

$
tv(i, j) =
\begin{cases}
 max \{ -A_{m,n} + 1 \ , 1 \} & \text{si } i = m  \ \land \ j =n \\ max \{ tv(i, j+1)-A_{i,j} \ , 1\} & \text{si } i = m\\ max \{ tv(i+1, j) -A_{i,j} \ , 1\} & \text{si } j= n\\ max \{ min \{ tv(i, j+1) , tv(i+1, j)\} - A_{i,j} \ , 1 \} & \text{caso contrario}
\end{cases}
$

```python
#%% Punto 16: Travesía Vital algoritmo dinámico Top down
m = 3
n = 3
M = [[float('inf') for _ in range(n)] for _ in range(m)]
A = [[-2, -3, 3], [-5,-10, 1], [10, 30, -5]]
def tv(i,j):

    #Caso información ya guardada
    if M[i][j] != float('inf'):
        return M[i][j]

    #Caso base: llegué al final del terreno
    if i == m-1 and j == n-1:
        celda = A[i][j]
        if celda >=0: #Si es positivo, con haber llegado con 1 de vida me alcanza
            return 1

        else: #Si es negativo, significa que debo haber llegado con uno más que el valor absoluto de mi celda
            return abs(celda) + 1


    #Vamos con casos recursivos

    mvn = 0
    celda_actual = A[i][j]
    izq_der_disponibles = (i < m-1 and j < n-1)

    if izq_der_disponibles:
        vida_necesaria_a_posteriori = min(tv(i+1, j), tv(i, j+1))
        #Tengo dos casos: estoy sobre una poción o sobre veneno
        #Si tengo pocion, y esta basta para soportar el resto del camino, entonces simplemente debo haber llegado con 1 de vida a ella.
        #Si tengo pocion y no me basta, debo de haber entrado con más vida, para poder, junto a la poción, soportar el veneno más tarde
        #Si tengo veneno, debo adicionarle a mi vida necesaria al entrar el valor absoluto del veneno
        #Luego:
        mvn = max(vida_necesaria_a_posteriori - celda_actual, 1)
    #Los casos en los que solo puedo moverme en una dirección son análogos, salvo que está determinado "cuál es mi futuro" (bajar o derecha)
    elif i == m-1:
        vida_necesaria_a_posteriori = tv(i,j+1)
        mvn = max(vida_necesaria_a_posteriori - celda_actual, 1)
    elif j == n-1:
        vida_necesaria_a_posteriori = tv(i+1,j)
        mvn = max(vida_necesaria_a_posteriori - celda_actual, 1)

    #Guardo valor calculado en mi estructura de memoizacion
    M[i][j] = mvn

    return mvn


print(f'La cantidad mínima de vida inicial necesaria es {tv(0,0)}')
```

Analicemos complejidad del algoritmo.

Por un lado, en cuanto a la complejidad espacial auxiliar, tengo simplemente una matriz (mi estructura de memoización) de tamaño $O(m*n)$.

Por otro lado, en cuanto a la complejidad temporal, tengo definidos en total $O(m*n)$ estados, puesto que las combinaciones de entradas $(i,j)$ con $0 \leq i \leq m-1 \ \land \ 0 \leq j \leq n-1$, generan que en conjunto haya $n*m$ posibles combinaciones.

Luego, en cada estado, las operaciones realizadas son sumas, comparaciones, selección de mínimo/máximo, consultas a mi matriz de memoización y asignaciones; las cuales en conjunto generan una complejidad $O(1)$

En definitiva, mi complejidad temporal final es $O(m*n) * O(1) = O(m*n)$

Ahora bien, ¿será posible mejorar la complejidad espacial auxiliar del algoritmo? Pensemos en una implementación _bottom up_ que parta del caso base, y extienda la solución hasta el momento inicial.

Nótese que no es necesario estar guardando cada variante $(i,j)$.

Supongamos que tengo más filas que columnas, es decir, $m \geq n$

Entonces, sea $M$ nuestra estructura de memoización un arreglo de tamaño $O(n)$. Luego, notar que:
- En el caso base $i=m, \ j = n$ entonces directamente establecemos en $M[j]$ el valor $max \{1, -A_{m,n} + 1 \}$.

- Si $i = m-1$, entonces en vez de consultar $M[i][j+1]$ como antes, es claro que _la columna derecha_ a la que debo consultar la información sobre la vida necesaria _a posteriori_ es $M[j+1]$ (cuyo valor ya fue establecido, por la dirección de la recursión mencionada). Luego, reescribo ahí mismo la nueva cantidad necesaria.

- Si $j = n-1$, entonces en vez de consultar $M[i+1][j]$ como antes, es claro que la _fila de abajo_ puede ser el mismo $M[j]$ (cuyo valor ya fue establecido, por la dirección de la recursión mencionada), y luego reescrbo ahí mismo la nueva cantidad necesaria.

- Si en cambio ambos movimientos están disponibles, entonces bastaría tomar $v = min(M[j], M[j+1]$ que representan abajo y derecha, respectivamente. Y luego, reemplazar en $M[j]$ el valor $max \{ v - A_{i,j}, 1\}$ como antes.


${}$



El caso $m < n$ es análogo

Es entonces que, de esta manera, logaríamos reducir la complejidad espacial auxiliar a $O(min(m,n))$

```python
#%% algoritmo dinámico bottom up
m = 3
n = 3
minimo = min(m,n)
M = []
M = [float('inf')for _ in range(minimo)]
A = [[-2, -3, 3], [-5,-10, 1], [10, 30, -5]]

def tvBU(A):

    m = len(A)
    n = len(A[0])
    mvn = 0
    if m>=n:
        for i in range(m-1, -1, -1):
            for j in range(n-1, -1, -1):
                celda_actual = A[i][j]
                if i == m-1 and j==n-1:
                    mvn = max(-celda_actual + 1, 1)
                    M[j] = mvn
                elif i == m-1:
                    mvn = max(M[j+1] - A[i][j], 1)
                    M[j] = mvn
                elif j == n-1:
                    mvn = max(M[j] - A[i][j], 1)
                    M[j] = mvn
                else:
                    v = min(M[j+1], M[j])
                    mvn = max(v - A[i][j], 1)
                    M[j] = mvn
    else:
        for j in range(n-1, -1, -1):
            for i in range(m-1, -1, -1):
                celda_actual = A[i][j]
                if i == m-1 and j==n-1:
                    mvn = max(-celda_actual + 1, 1)
                    M[i] = mvn
                elif i == m-1:
                    mvn = max(M[i] - A[i][j], 1)
                    M[i] = mvn
                elif j == n-1:
                    mvn = max(M[i+1] - A[i][j], 1)
                    M[i] = mvn
                else:
                    v = min(M[i], M[i+1])
                    mvn = max(v - A[i][j], 1)
                    M[i] = mvn

    return M[0]


test1 = [[1, -3, 3], [0, -2, 0], [-3, -3, -3]]
test2 = [[-5, 2], [1, -1], [-10, -2], [2, -3]]
test3 = [[-2, -5, 10, -1], [10, -1, -5, -2]]

print(f"Test Cuadrado (Esperado 3): {tvBU(test1)}")
print(f"Test Alto (Esperado 10): {tvBU(test2)}")
print(f"Test Ancho (Esperado 3): {tvBU(test3)}")
```

#Punto 17: PilaCauta modificada

Tenemos cajas numeradas de $1$ a $N$, todas de iguales dimensiones. Queremos encontrar la cantidad de formas distintas que pueden apilarse las cajas en una única pila cumpliendo que:
 - sólo puede haber una caja apoyada directamente sobre otra;
 - las cajas de la pila deben estar ordenadas crecientemente por número, de abajo para arriba;
 - cada caja $i$ tiene un peso $w_i$ y un soporte $s_i$, y el peso total de las cajas que están arriba de otra no debe exceder el soporte de esa otra.

Si tenemos los pesos $w = [19,7,5,6,1]$ y los soportes $s = [15,13,7,8,2]$ (la caja 1 tiene peso 19 y soporte 15, la caja 2 tiene peso 7 y soporte 13, etc.), entonces la respuesta es 29. Por ejemplo, pueden apilarse de la forma 1-2-3-5 o 1-2-4-5 (donde la izquierda es más abajo), entre otras opciones.

Sea $pc(s_p, i)$ la función que resuelve el problema, cuyos parámetros son una solución parcial y un iterador de las cajas. Notar que en cada oportunidad, la decisión a tomar es si agregar o no la caja a la pila. Es decir, dada una solución parcial $s_p$ y un iterador $i$, luego:
- O bien $i=0$ entonces significa que no agregué todavía ningún elemento, y por lo tanto $s_p = [ \ ]$ . Entonces puedo agregar este elemento y ya tener una solución válida, puedo agregar el elemento y buscar más soluciones óptimas o puedo no agregar el elemento y buscar soluciones que inicien con otra caja. Así: $res = 1 + pc(s_p \oplus \{i\},i+1) + ps(s_p, i+1)$.
- O bien $ N > i \neq 0$ pero $s_p = [ \ ]$, es decir, no agregamos los elementos $0, ..., i-1$, y así nuevamente tengo una solución válida, y puedo o bien agregar la i-ésima caja y buscar más soluciones, o bien no agregar la i-ésima caja y seguir buscando otra caja como base. Es decir, $res = 1 + pc(s_p \oplus \{ i\}, i+1) + pc(s_p, i+1)$
- O bien $s_p \neq [ \ ] \ $ e $\ N < i \neq 0$ entonces:
  
  - Si $\forall 0\leq j < len(s_p) \ (\sum_{k= 0}^{j} w_{s_p[k]}) + w_i< s_j$ entonces significa que puedo adicionar a mi pila la i-ésima caja y tener una nueva solucion válida, y luego continuar adicionando otras; o no agregarla, y buscar más soluciones. Así $res = 1 + pc(s_p \oplus \{i \}, i+1) + pc(s_p, i+1)$

  - En caso contrario, no puedo adicionar la i-ésima caja, y simplemente la paso de largo; luego $res = pc(s_p, i+1)$

Ahora bien, pensemos una nueva idea, pero que esta vez la recursión esté orientada hacia la base de un algoritmo de programación dinámica.

¿Qué requiero conocer yo en todo momento?

Es claro que es necesario conocer el elemento (o, en la práctica, el iterador) $i$. Pero, ¿es necesario estar pasando como parámetro de nuestra función cada solución parcial?

Notar que lo importante de una solución parcial es si puede o no soportar la i-ésima caja. Ahora bien, es necesario que cada caja de la solución parcial soporte la que están encima. Luego, eso signficaría que, en el peor de los casos, deba estar recordando la capacidad de soporte "disponible" (es decir, la capacidad que le sobra) de cada una, y eso podrían ser hasta $N$ parámetros(o un arreglo de tamaño $O(N)$). Pero, ¿y si en cambio solo recordamos el mínimo? Es decir, que la función sea $pc(i, c)$ donde $i$ representa la i-ésima caja a considerar, y $c$ representa la mínima capacidad restante a respetar. Luego, si al intentar agregar la i-ésima caja, ocurre que se supera ese mínimo, significaría que en algún punto de la pila, una caja está soportando un peso mayor a su nivel de soporte, y por lo tanto ocurre un derrumbe.

Planteemos esta idea formalmente:

$
pc(i, c) =
\begin{cases}
 1 + pc(i+1, s_i) + pc(i+1, 0) & \text{si } i < N  \ \land \ c = 0 \\ 1 + pc(i+1, min \{c-w_i, s_i \}) + pc(i+1, c) & \text{si } i < N \ \land \ c > w_i \\ pc(i+1, c) & \text{si } i < N \ \land c \leq w_i \\ 0 & \text{caso contrario}
\end{cases}
$


Luego, si llamamos $c_M$ a la capacidad máxima de soporte entre las cajas, entonces es posible armar una estructura de memoización que sea una matriz $M$ de dimensiones $N \times c_M$, es decir, el espacio auxiliar presente una complejidad $O(N *c_M)$ ; que representaría la cantidad de estados posibles en el peor caso.

```python
#Algoritmo dinámico Top Down

N = 5
w= [19,7,5,6,1]
s = [15,13,7,8,2]
cM = max(s)
M = [[float('inf') for _ in range(cM+2)] for _ in range (N+1)]
def pc(i, c):

    #Caso info ya guardada
    if M[i][c] != float('inf'):
        return M[i][c]

    #Caso todas las cajas ya revisadas
    if i == N:
        return 0

    #Caso recursivo

    if c == 0: #Caso no se agregó ninguna caja a la pila
        M[i][c] = 1 + pc(i+1, s[i]+1) + pc(i+1, 0) #Agrego "+1" a la capacidad si agrego la caja, para que entonces no permitir que pueda volver a ser 0 (que es caso vacío)
        return M[i][c]
    elif c - w[i] <= 0: #caso pila no soporta caja
        M[i][c] = pc(i+1, c)
        return M[i][c]
    else: #caso pila soporta caja
        M[i][c] = 1 + pc(i+1, min(c - w[i], s[i] + 1)) + pc(i+1, c)
        return M[i][c]

print(pc(0, 0))
```

Ahora bien, ¿y si mejor nuestra estructura de memoización recuerda cuántas formas distintas de apilar hay por soporte mínimo considerado? Es decir, si dos pilas comparten mismo soporte mínimo, ¿por qué no llevar la cuenta de las distintas formas a considerarse bajo una misma celda/espacio en nuestra memoria?

Bajo esta lógica, presento un algoritmo bottom up, que mejora la complejidad auxiliar espacial a $O(C_M)$

```python
w= [19,7,5,6,1]
s = [15,13,7,8,2]

def pc(N):
    M = {} #creo la estructura de memoización vacía

    for i in range(N): #recorro las cajas
        dict_aux = M.copy() #creo una estructura auxiliar para guardar los resultados que se irán generando, para después pasarlos a la estructura original
        if s[i] in dict_aux:  #si el soporte ya fue considerado
            dict_aux[s[i]] += 1
        else: #si el soporte mínimo no fue analizado aún:
            dict_aux[s[i]] = 1
        for (soporte, cant_formas) in M.items(): #recorramos los soportes que tenemos en nuestros diccionarios, para ver si podemos crear nuevas pilas adicionando la i-ésima caja
            if soporte - w[i] >= 0: #caso apilable
                soporte_minimo = min(soporte - w[i], s[i]) #defino el mínimo soporte que tiene la pila
                #Tener en cuenta que, al poder apilar la caja, tengo cant_formas de nuevas pilas; porque ese número representa la cantidad de pilas que poseen el soporte inicial
                if soporte_minimo in dict_aux:
                    dict_aux[soporte_minimo] += cant_formas
                else:
                    dict_aux[soporte_minimo] = cant_formas
        M = dict_aux.copy() #paso la información a mi anterior diccionario

    #Por último, debo contar la cantidad de formas totales halladas
    res = 0
    for clave in M:
        res += M[clave]

    return res

print(pc(5))
```

En ambos casos, sin embargo, la complejidad temporal es $O(N*c_M)$, puesto que representa la cantidad de estados posibles en el peor caso; sumado a que en cada estado las operaciones que se realizan son $O(1)$

#Punto 18: OperacionesSeq

Sea $v = (v_1,v_2,...v_n)$ un vector de números naturales, y sea $w \in \mathbb{N}$. Se desea intercalar entre los elementos de $v$ las operaciones $+$ (suma), $×$ (multiplicación) y $↑$ (potenciación) de tal manera que al evaluar la expresión obtenida el resultado sea $w$. Para evaluar la expresión se opera de izquierda a derecha ignorando la precedencia de los operadores. Por ejemplo, si $v = (3,1,5,2,1)$, y las operaciones elegidas son $+$, $×$, $↑$ y $×$ (en ese orden), la expresión obtenida es $3+1×5 ↑ 2×1$,
que se evalúa como $(((3 +1)×5) ↑ 2)×1 = 400$.

Escribir una formulación recursiva que, dados $v$ y $w$, encuentre una secuencia de operaciones como la deseada, en caso de que tal secuencia exista.

Sea $v \in \mathbb{N}^n, w \in \mathbb{N}$.

En principio, podemos averiguar si es que realmente existe una solución, y en caso de que así sea, reconstruirla.

Pensemos lo siguiente: armar una función que se pregunte ¿es posible alcanzar este número?

Notar que, para ello, la función debería recorrer el vector $v$ en reversa. Es decir, arrancamos preguntando ¿es posible llegar a $w$?. Y la respuesta será si hubo forma de alcanzar $w'_1 = w - v_n \ \lor \ w'_2 = w \div v_n \ \lor \ w'_3 = \sqrt[v_n]{w} $

Y, la respuesta será _no_ si $w < 0 \ \lor \ w \neq \lfloor w \rfloor \ \lor \ (i=1 \ \land w>0) $

Es decir, o bien no tengo un número natural, o bien alcancé el inicio de mi vector y no pude conseguir formar a $w$ (es decir, no llegué a conseguir $w' = v_1$)


Luego, se tiene que la función quedaría:

$
\text{os}(i, w') =
\begin{cases} False
 & \text{si } w' < 0 \ \lor \ w' \neq \lfloor w' \rfloor \ \lor \ (i = 1 \ \land \ w' \neq v_1) \\ True & \text{si } i = 1 \ \land \ w' = v_1\\ os(i-1, w'-v_i) \lor os(i-1, w' \div v_i)  \lor os(i-1, \sqrt[v_i]{w'})  & \text{sino }
\end{cases}
$


Y entonces, la respuesta a nuestro problema es $os(n, w)$

Luego, para reconstruir la solución en cado de que $os(n,w)$ sea $True$, puedo ir consultando mi matriz (en función si la resta, división y/o raíz fueron válidas) e ir agregando los símbolos siempre que tenga un $True$ en el paso anterior con el respectivo $w'$.

Tener en cuenta, sin embargo, que para un algoritmo dinámico, conviene antes de llamar a $os(i-1, w' - v_i)$, $os(i-1, w' \div v_i)$ y $os(i-1, \sqrt[v_i]{w'})$, verficar si son naturales $w' - v_i$ ; $w' \div v_i$ ; $\sqrt[v_i]{w'}$ respectivamente.

Adicionalmente, en vez de reconstruir la solución preguntando booleanos, puedo guardar qué operación se hubo realizado para alcanzar cada $w'$.

```python
# Algoritmo dinámico Top Down
v = [3,1,5,2,1]
w = 400
M = [[None for _ in range(w+1)] for _ in range(len(v))]

def os(i, w_prima):

    if w_prima < 0 or (i == 0 and w_prima != v[0]):
        return False

    if M[i][w_prima] != None:
        return M[i][w_prima] != 'nada'

    if i == 0 and w_prima == v[0]:
        M[i][w_prima] = True
        return True

    res_general = False
    simbolo_alcanzado = 'nada'
    es_natural_la_resta = (w_prima - v[i] >= 0)
    es_divisible = (w_prima % v[i] == 0)
    raiz_cand = round(w_prima**(1/v[i]))
    es_natural_la_raiz = (raiz_cand**v[i] == w_prima)

    if es_natural_la_resta:
        res_resta = os(i-1, w_prima - v[i])
        res_general = res_resta
        if res_resta is True:
            simbolo_alcanzado = '+'
    if es_divisible and not(res_general):
        res_division = os(i-1, w_prima // v[i])
        res_general = res_general or res_division
        if res_division is True:
            simbolo_alcanzado = 'x'
    if es_natural_la_raiz and not(res_general):
        res_raiz = os(i-1, raiz_cand)
        res_general = res_general or res_raiz
        if res_raiz is True:
            simbolo_alcanzado = '↑'

    M[i][w_prima] = simbolo_alcanzado

    return res_general

existe_sol = os(len(v) - 1, 400)
print(existe_sol)
#Reconstruyamos la solución
if existe_sol:
    res_reverso = []
    res_real = []
    w_prima = w
    for i in range(len(v) - 1, 0, -1):
        res_reverso.append(M[i][w_prima])
        simbolo = M[i][w_prima]
        if simbolo == '+':
            w_prima = w_prima - v[i]
        elif simbolo == 'x':
            w_prima = w_prima // v[i]
        else:
            w_prima = round(w_prima**(1/v[i]))

    for j in range(len(res_reverso) - 1, -1, -1):
        res_real.append(res_reverso[j])

    print(res_real)
```
