#Punto 27: SumaSelectiva

Dado un conjunto $X$ con $|X| = n$ y un entero $k \leq n$ queremos encontrar el máximo valor que pueden sumar los elementos de un subconjunto $S$ de $X$ de tamaño $k$. Más formalmente, queremos calcular $\underset{S⊆X , \ |S|=k}{máx} \sum_{s∈S} s$

Para resolver el problema, puedo usar un algoritmo de ordenamiento sobre $X = \{x_1,...,x_n \}$, de forma tal de obtener $X' = \{ x_{i_1},...,x_{i_n} \}$ tal que para cada $1 \leq j < n$ se tenga que $x_{i_j} \leq x_{i_{j+1}}$.
Y luego, $\underset{S⊆X , \ |S|=k}{máx} \sum_{s∈S} s = \sum_{j=1}^k x_{i_j} $


Pseudocódigo:

Entrada: arreglo $X$, entero $k \leq len(X)$

Salida: $\underset{S⊆X , \ |S|=k}{máx} \sum_{s∈S} s$


$X\_prima$ <- _ordenar($X$, decreciente)_ \\\\ en pricipio puedo usar cualquier algoritmo de sorting

$S$ <- $[ \ ]$

valor_maximo <- $0$

Para $i$ en rango $k$:

$\hspace{1cm}$  $S$ <- $S \cup \{X\_prima[i] \}$

$\hspace{1cm}$ valor_maximo <- valor_maximo + $X\_prima[i]$

Retornar valor_maximo, $S$

**Demostremos su correctitud**

Veamos que el algoritmo devuelve el subconjunto $S$ que contiene aquellos elementos con los que se obtiene el máximo valor alcanzable en la suma de $k$ elementos de $X$.

Teorema: el algoritmo greedy devuelve la solución $S$ óptima al problema _SumaSelectiva_.

$ \\ $

$ \\ $

Lema (subestructura óptima):

Sea $\mathcal{G} = \{g_1,...,g_k \}$ la solución greedy al problema de _SumaSelectiva_. Luego, para todo $i < k$,$\quad g_{i+1} = max \ x \in X \setminus \{g_1,...g_i \}$

$  $

Demostración del lema por inducción:

Caso base $i = 0$:

$g_{0+1} = g_1 \underset{def}{=} max \ x \in X = max \ x \in X \setminus \{ \} \quad \checkmark$

Hipótesis inductiva: supongamos que $g_i = max \ x \in X \setminus \{g_1,...,g_{i-1} \} \quad$ para $1 < i < k$

Paso inductivo: supongamos que no es cierto, es decir, existe $x* \in X \setminus \{g_1,...,g_{i-1} \}$ tal que $x*>g_{i+1}$

Luego, dado que $x* > g_{i+1}$ entonces por la definición del algoritmo, $\exists 1 \leq j \leq i : g_j = x*$. Pero luego, como $j \leq i < i+1$, entonces $x* \in \{g_1,...,g_i \} \Rightarrow x* \notin X \setminus \{g_1,...,g_{i-1} \} \quad$

O sea $x* \in X \setminus \{g_1,...,g_{i-1} \} \ \land x* \notin X \setminus \{g_1,...,g_{i-1} \} \quad ABS$

Por lo tanto, $x* \leq g_{i+1} \ \ \forall x* \in X \setminus \{g_1,...,g_{i-1} \} \Rightarrow g_{i+1} = max \ x \in X \setminus \{g_1,...,g_{i-1} \} \quad \checkmark$


Finalmente, por principio de inducción, se cumple el lema.

$\hspace{17cm} \huge\boxtimes$

$ $

Lema principal: _Greedy Stays Ahead_

Sea $O = \{o_1,...,o_k \}$ una solución óptima al problema de _SumaSelectiva_, ordenada de forma decreciente; y sea $\mathcal{G} = \{g_1,...,g_k\}$ la solución greedy.

Luego $o_i \leq g_i \quad \forall \ 1\leq i\leq k$

$ $

Demostración del lema: por inducción en $i$

Caso base $i=1$ :

Por construcción, $g_1 = max \{x \in X\} \geq o_1 \quad \checkmark $

Hipótesis inductiva: se cumple que $o_j \leq g_j $ para todo $j \leq i < k $

Objetivo: ver que $o_{i+1} \leq g_{i+1}$

Paso inductivo:

Notar que $o_{i+1} < o_i \leq g_j \quad \forall j \leq i$

Luego, $o_{i+1} \in X \setminus \{ g_1,...,g_i \}$

Y, por el lema anterior, $g_{i+1} = max \ x \in X \setminus \{ g_1,...,g_i \}$

Luego, se deduce que $g_{i+1} \geq o_{i+1} \quad \checkmark$


$ $

Demostremos entonces el teorema:

Sea $O = \{ o_1,...,o_k \} $ una solución óptima ordenada de forma decreciente, y sea $\mathcal{G} = \{ g_1, ..., g_k\}$ la solución greedy

Luego, por el lema _Greedy Stays Ahead_, se tiene que $o_i \leq g_i \quad \forall 1 \leq i \leq k$

Luego, $\sum_{o \in O} o = \sum_{i=1}^k o_i \ \leq \sum_{i=1}^k g_k = \sum_{g \in \mathcal{G}} g$

Pero, como $O$ solución óptima, entonces $\sum_{o \in O} o = \sum_{g \in \mathcal{G}} g$

Y luego $\mathcal{G}$ es una solución óptima.

$\hspace{17cm} \huge\boxtimes$

```python
#%% suma selectiva en O(n log n)

# Primero usaremos merge_sort para ordenar X
# El algoritmo lo obtuve de asimov.cloud

def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    # 1️⃣ División
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    # 2️⃣ Fusión
    return merge(left, right)


def merge(left, right):
    merged = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(right[j])
            j += 1
        else:
            merged.append(left[i])
            i += 1
    # Añadir los restos (solo uno de los dos bucles se ejecutará)
    merged.extend(right[j:])
    merged.extend(left[i:])
    return merged

def sumaSelectiva(X, k):
    X_ordenada = merge_sort(X)
    S = X_ordenada[:k]
    valor_maximo = sum(X_ordenada[i] for i in range(k))

    return valor_maximo, S

print(sumaSelectiva([1000, 4,2,7,9,100,1], 3))
```

```python
#%%suma selectiva en O(n log k)

def k_mas_grandes(X, k):
    res = []
    for i in range(len(X)):
        if i < k-1:
            res.append(X[i])

        if i == k -1:
            res.append(X[i])
            build_heap(res)

        if i>= k:
            if res:
                res = min_heap(res, X[i])

    return res

def build_heap(arr):
    k = len(arr)
    i = k // 2 - 1
    while i > -1:
        heapify(arr, i)
        i -= 1



def min_heap(res, e):
    if res[0] < e:
        res[0] = e
        heapify(res, 0)

    return res

def heapify(arr, i):
    k =len(arr)
    smallest = i          # Inicializa el mayor como raíz
    left = 2 * i + 1     # hijo izquierdo
    right = 2 * i + 2    # hijo derecho

    # Si el hijo derecho es menor que el menor actual
    if right < k and arr[right] < arr[smallest]:
        smallest = right

    # Si el hijo izquierdo es menor que la raíz
    if left < k and arr[left] < arr[smallest]:
        smallest = left


    # Si el menor no es la raíz, intercambiamos y continuamos heapificando
    if smallest != i:
        arr[i], arr[smallest] = arr[smallest], arr[i]
        heapify(arr, smallest)




def sumaSelectivaV2(X,k):
    k_mayores = k_mas_grandes(X,k)
    X_ordenada = merge_sort(k_mayores)
    S = X_ordenada[:k]
    valor_maximo = sum(X_ordenada[i] for i in range(k))

    return valor_maximo, S

print(sumaSelectivaV2([1000, 4,2,7,9,100,1], 3))
```

# Punto 28: SumaGolosa

Queremos encontrar la suma de los elementos de un multiconjunto de números naturales. Cada suma se realiza exactamente entre dos números $x$ e $y$ y tiene costo $x + y$.

Por ejemplo, si queremos encontrar la suma de ${1,2,5}$ tenemos 3 opciones:

 - $1 + 2$ (con costo $3$) y luego $3 + 5$ (con costo $8$), resultando en un costo total de $11$;

- $1 + 5$ (con costo $6$) y luego $6 + 2$ (con costo $8$), resultando en un costo total de $14$;

- $2 + 5$ (con costo $7$) y luego $7 + 1$ (con costo $8$), resultando en un costo total de $15$.

Queremos encontrar la forma de sumar que tenga costo mínimo, por lo que en nuestro ejemplo la mejor forma sería la primera.

Esta es la idea: sea $X = \{x_1, x_2,...,x_n\}$ los elementos a sumar.

Lo primero que hacemos es ordenarlo.

Sea $X_{ord_1} = ordenar(X, creciente)$. Luego, si $X_{ord_1} = \{ x_1^*, x_2^*, ..., x_n^*\}$ entonces se cumple que $x_i^* \leq x_j^* \quad \forall 1 \leq i < j \leq n$

Luego, sea $s_1 = x_1^* + x_2^*$. Nótese que $s_1$ representa la suma de menor costo posible que se puede hacer entre dos elementos de $X_{ord_1}$ (o de $X$ mismo, ya que una es una permutación del otro).

Entonces, lo restante que queda por sumar es $\{ s_1, x_3^*, ..., x_n^*\}$

Nuevamente, repetimos proceso: sea $X_{ord_2} = ordenar(\{ s_1, x_3^*, ..., x_n^*\}, creciente)$

Si tomamos $X_{ord_2} = \{ x'_1, x'_2,...,x'_n \}$ entonces definimos $s_2 = x'_1 + x'_2$

Luego, $s_2$ representa la mínima suma que se puede realizar después de haber realizado una suma anteriormente.


El algoritmo greedy implementa esta estrategia hasta que ya no halla elementos por sumar.

Algoritmo

Entrada $X = [x_1,...,x_n]$

$X_{ord}$ <- $ordenar(X)$


costo <- 0

Hasta que $k = 1$ hacer:

$\hspace{1cm}$  suma <- $X_{ord}[1] + X_{ord}[2]$

$\hspace{1cm}$ elementos_restantes_por_sumar <- $ \{ $ suma $ \} \ \cup \ X_{ord}[3:]$

$\hspace{1cm}$ costo <- costo + suma

$\hspace{1cm}$ k <- len(elementos_a_sumar)

$\hspace{1cm} X_{ord}$ <- ordenar(elementos_a_sumar)

$ $

Devolver costo

Demostremos que la estrategia propuesta resuelve el problema

$Teorema:$ la estrategia greedy implementa la forma de sumar que tiene costo mínimo.

$ $

Empecemos antes demostrando los siguientes lemas auxiliares:

$Lema_1 :$ sea $X$ el conjunto de elementos a sumar, con $n = len(X)$, entonces se deben realizar $n-1$ sumas.

$Dem_1:$

Sea $X$ un conjunto de elementos a sumar, de tamaño $n$

Luego, por cada suma $s$ realizada en $X$, la cantidad de elementos por sumar se reduce en una unidad. Es decir, para la suma $s$ se requieren dos elementos, y se devuelve un único elmento. Luego, quedan por sumar $n-1$ elementos.

Por lo tanto, si en cada suma se reduce en una unidad la cantidad total de elementos por sumar, y el proceso termina cuando se alcanza $k=1$, entonces, la cantidad necesaria de sumas es $x$ tal que $n - x = 1 \ \iff x=n-1$, como se quería ver.

$\hspace{23cm} \huge\boxtimes$


$Lema_2$: sea $\mathcal{G} = \{ g_1,...,g_n\}$ la secuencia de sumas de la solución greedy para el problema de _sumaGolosa_ para un conjunto de tamaño $n+1$. Luego, existe solución óptima $\mathcal{O} = \{g_1, o_2,...,o_n \}$ (es decir, existe una solución óptima que empieza realizando la suma que dicta la solución greedy)

$Dem_2:$

Sea $X = \{x_1,...,x_{n+1}\}$

Nótese que, para una solución cualquiera, $costo = \sum_{i=1}^{n+1} x_i d_i$ donde $d_i$ es la cantidad de veces en la que el i-ésimo elemento participa en la suma ("es sumado").

Sean luego $x_{n_0}, x_{n_1}$ tales que $x_{n_0}, x_{n_1} \leq x_i \quad \forall  \ 1 \leq i \leq n+1 \ \land \ (i \neq n_0, n_1)$

Sea entonces $\mathcal{O} = \{o_1,...,o_n \}$ la secuencia (en orden) de sumas de una solución óptima tal que $o_1 = x_{k_0} + x_{k_1}$ con $(x_{k_0} , x_{k_1}) \neq (x_{n_0}, x_{n_1})$

Y sea $o_{n_0}$ la primer suma en la que está involucrada $x_{n_0}$, es decir, tal que $o_{n_0} = x_{n_0} + s \ $ y $ \ x_{n_0} \notin o_i \quad \forall 1 \leq i < n_0$

De manera análoga, sea $o_{n_1}$ la primer suma en la que está involucrada $x_{n_1}$, es decir, tal que $o_{n_1} = x_{n_1} + s' \ $ y $\ x_{n_1} \notin o_i \quad \forall 1 \leq i < n_1$

Luego, si tomamos $\mathcal{O'} = \{ g_1,o_2..., o_{n_0 -1}, o_{n_0} + s,..., o_{n_1 - 1}, x_{k_1} + s',..., o_n\}$

Es decir, la diferencia entre $\mathcal{O}$ y $\mathcal{O'}$ desde el punto de vista de costo es que, si tomamos $costo_{\mathcal{O}} = \sum_{i=1}^{n+1} x_i d_i$, luego, si suponemos SPG $n_0 < n_1 < k_0 < k_1$:

$costo_{\mathcal{O'}} = (\sum_{i=1}^{n_0-1} x_i d_i )+ x_{n_0}*d_{k_0} + (\sum_{i=n_0+1}^{n_1-1} x_i d_i) + x_{n_1}*d_{k_1} + (\sum_{i=n_1 + 1}^{k_0-1} x_i d_i) + x_{k_0}*d_{n_0} + (\sum_{i=k_0 + 1}^{k_1-1} x_i d_i) + x_{k_1}*d_{n_1} + \sum_{i=k_1+1}^{n+1} x_i d_i$


Es decir, $costo_{\mathcal{O}} - costo_{\mathcal{O'}} = x_{n_0}*d_{n_0} - x_{n_0}*d_{k_0} + x_{n_1}*d_{n_1} - x_{n_1}*d_{k_1} + x_{k_0}*d_{k_0} - x_{k_0}*d_{n_0}  + x_{k_1}*d_{k_1} - x_{k_1}*d_{k_1}$

$costo_{\mathcal{O}} - costo_{\mathcal{O'}} = (x_{n_0} \cdot d_{n_0} + x_{k_0} \cdot d_{k_0}) - (x_{n_0} \cdot d_{k_0} + x_{k_0} \cdot d_{n_0}) + (x_{n_1} \cdot d_{n_1} + x_{k_1} \cdot d_{k_1}) - (x_{n_1} \cdot d_{k_1} + x_{k_1} \cdot d_{n_1})$

$ costo_{\mathcal{O}} - costo_{\mathcal{O'}} = (x_{k_0} - x_{n_0})(d_{k_1} - d_{n_0}) + (x_{k_1} - x_{n_1})(d_{k_1} - d_{n_1})$

Donde cada paréntesis es positivo o cero, y luego $costo_{\mathcal{O}} \geq costo_{\mathcal{O'}}$

Pero como $\mathcal{O}$ es solución óptima, entonces $costo_{\mathcal{O}} = costo_{\mathcal{O'}}$

Y por lo tanto, $\mathcal{O'}$ también lo es.

$\hspace{25cm} \huge\boxtimes$

De esta demostración, se desprende además que la solución óptima de un conjunto de tamaño $n+1$ equivale a la suma de los menores elementos más la solución óptima del conjunto de tamaño $n$ que contiene a los elementos restantes junto con el resultado de la suma descripta (ver $\mathcal{O'})$.


$ $

Luego, el teorema se demuestra por inducción en $n$ fácilmente, puesto que los casos base $n= 0$ y $n=1$ son triviales, y el paso inductivo sale que si se cumple para $k < n+1$, la solución óptima será equivalente a la suma de los dos menores más la solución óptima del conjunto de tamaño $n$ que contiene a esta suma junto con el resto de los elementos (que, por $HI$, es lo que devuelve el algoritmo greedy para un conjunto de tamaño $n$), y luego, en conjunto, esto equivale además a la solución greedy para un conjunto de tamaño $n+1$.



$\hspace{25cm} \huge\boxtimes$

```python
#Implementación algoritmo suma golosa en O(nlog n) usando min heap

def build_heap(arr):
    k = len(arr)
    i = k // 2 - 1
    while i > -1:
        heapify(arr, i)
        i -= 1



def heapify(arr, i):
    k =len(arr)
    smallest = i          # Inicializa el mayor como raíz
    left = 2 * i + 1     # hijo izquierdo
    right = 2 * i + 2    # hijo derecho

    # Si el hijo izquierdo es menor que la raíz
    if left < k and arr[left] < arr[smallest]:
        smallest = left

    # Si el hijo derecho es menor que el menor actual
    if right < k and arr[right] < arr[smallest]:
        smallest = right


    # Si el menor no es la raíz, intercambiamos y continuamos heapificando
    if smallest != i:
        arr[i], arr[smallest] = arr[smallest], arr[i]
        heapify(arr, smallest)

def extraer_minimo(X):
    minimo = X[0]
    X[0] = X[len(X)-1]
    X.pop()
    heapify(X, 0)
    return minimo

def heapifyUp(X, i):
    if i>0:
        padre = (i-1)//2
        if X[padre] > X[i]:
            e = X[padre]
            X[padre] = X[i]
            X[i] = e
            heapifyUp(X, padre)


def agregar_elemento(X, e):
    X.append(e)
    heapifyUp(X, len(X) -1)

def suma_golosa(X):
    X_ord = X.copy()
    build_heap(X_ord)

    costo = 0
    k = len(X_ord)

    while k != 1:
        elem_mas_chico = extraer_minimo(X_ord)
        seg_elem_mas_chico = extraer_minimo(X_ord)
        suma = elem_mas_chico + seg_elem_mas_chico
        costo += suma
        agregar_elemento(X_ord, suma)
        k = len(X_ord)

    return costo


print(suma_golosa([2, 5, 8, 10]))
```

# Punto 32: División Pandémica

En medio de una pandemia, la Escuela de Aulas Grandes y Ventiladas quiere implementar un protocolo especial de distanciamiento social que tenga en cuenta que la escuela no tienen restricciones de espacio. El objetivo es separar a cada curso en dos subcursos a fin de reducir la cantidad de pares de estudiantes que sean muy cercanos, dado que se estima que estos estudiantes tienen dificultades para respetar tan buscado distanciamiento. Para este fin, en el protocolo se estableció que cada curso que tenga $c$ parejas de estudiantes cercanos tiene que dividirse en dos subcursos, cada uno de los cuales puede tener a lo sumo $\frac{c}{2}$ parejas de estudiantes cercanos. Notar que no importa si un subcurso queda con más estudiantes que otro.

Formalmente, para cada curso contamos con un conjunto de estudiantes $E$ y su conjunto $C$ de pares de estudiantes cercanos. Luego, una partición $(A,B)$ de $E$ es una solución factible para $(E,C)$ cuando $|(A×A)∩C| ≤ |C|/2$ y $|(B ×B)∩C| ≤ |C|/2$. Por ejemplo, si $E = \{1,2,3,4\}$ y $C = \{1-2,2-3,3-4\}$, entonces $(\{1,3,4\},\{2\})$ y $(\{2,4\},\{1,3\})$ son soluciones factibles.

En este problema, yo recibo como parámetros una lista de estudiantes $E$ y luego un conjunto $C \subset E \times E$ (notar que $C \neq E \times E$ pues es absurdo considerar los pares $(i,i)$ con $i \in E$, así como lo es contener el par $(k,j)$ y el $(j,k)$ simultáneamente para $k,j \in E$).

A partir de esta información, el problema pide que devuelva particiones $A, B \subseteq E$ de los estudiantes, con $A \cap B = \emptyset $ (puesto que un estudiane no puede estar en dos cursos) tal que en cada partición haya como mucho la mitad de pares de estudiantes cercanos. Es decir, dado que $C$ es el conjunto de pares de estudiantes cercanos, luego $|C|$ es la cantidad total de pares de estudiantes cercanos; y busco que haya como máximo $|C|/2$ pares de estudiantes cercanos en $A$, $B$.

Es decir, si tomamos $A \times A$, $B \times B$ las posibles parejas de estudiantes en las particiones $A$ y $B$ respectivamente, entonces busco que

$|(A \times A) \ \cap \ C| \leq |C|/2$

$|(B \times B) \ \cap \ C| \leq |B|/2$


Nótese que $A \cup B = E$ y, además, $(C \cap (A \times A)) \cup (C \cap (B \times B)) \subset C$ pero no necesariamente $(C \cap (A \times A)) \cup (C \cap (B \times B)) = C$

Ahora bien, ¿es posible dado un conjunto de estudiantes $E$ y un conjunto de pares de estudiantes cercanos $C \subset E \times E$, existan particiones $A, B$ que cumplan lo pedido?

Demostremos que sí. ¿Cómo? Por inducción en $n=len(E)$


$ $

**Caso base $n=1$**: trivial, dado que $E = \{ 1 \}$, y, como vimos, $C \subsetneq E \times E = \{ (1,1) \}  $ entonces  $C = \emptyset$ y es por eso que me basta tomar $A = \{1\}, B= \emptyset$ puesto que $C \ \cap \ (A \times A) = C \ \cap \ (B \times B) = \emptyset$ y luego $|(A \times A) \ \cap \ C| = |(B \times B) \ \cap \ C| = |C|/2 = 0 \quad \checkmark$


**Caso base $n=2$**: trivial, dado que $E = \{ 1,2 \}$, y, como vimos, $C \subsetneq E \times E = \{ (1,1), (2,2), (1,2)\} $ (nota: consideramos $(1,2) = (2,1)$) entonces  $C = \emptyset \ $ (cuya posible partición puede ser $A = E, B = \emptyset$ por lo mismo que antes) o $\ C = \{ (1,2)\}$ y es por eso que me basta tomar $A = \{1\}, B= \{ 2\}$ puesto que $C \ \cap \ (A \times A) = C \ \cap \ (B \times B) = \emptyset$ y luego $|(A \times A) \ \cap \ C| = |(B \times B) \ \cap \ C| = 0 \leq |C|/2 = 1/2 \quad \checkmark$

**Hipótesis Inductiva**: supongamos que para todo conjunto de estudiantes $E$ de tamaño $j < n$ y un conjunto de pares de estudiantes $C \subsetneq E \times E$ se cumple que existe una partición $A, B \subseteq E$ tal que

$
\begin{aligned}
&|(A \times A) \cap C| \leq \tfrac{|C|}{2} \\
&|(B \times B) \cap C| \leq \tfrac{|C|}{2}
\end{aligned}
\quad
\left.
\begin{array}{l}
\\ \\   % estas líneas vacías fuerzan la altura
\end{array}
\right\}
\ast
$




**Objetivo**: ver que para un conjunto de estudiantes $E$ de tamaño $n$ y un conjunto de estudiantes $C \subsetneq E \times E$ se cumple que existen particiones $A, B \subseteq E$ que cumplen $\ast$.


**Paso Inductivo**: sean $E$ de tamaño $n$, $\ C \subsetneq E \times E$.

Luego, sea $E' = E \setminus \{1\}, \ C' = C \cap (E' \times E')$

Por hipótesis inductiva, como $len(E') = n-1 < n \ $ y $\ C' \subsetneq (E' \times E')$, entonces se cumple que existen $A', B' \subset E'$ tal que:

- $A' \cup B' = E' \ \land \ A' \cap B' = \emptyset$

- Se cumple $\ast$ si tomamos $C'$ como el conjunto de pares de estudiantes


Luego, sea $a = |C \cap ((A' \times {1}) \cup ({1} \times A'))|, \  b = |C \cap ((B' \times {1}) \cup ({1} \times B'))|$

Nótese que $|C| = |C'| + a + b$

Entonces, si tomo al mínimo entre $a, b$ (SPG, supongamos que es $a$; el otro caso es simétrico) entonces

$|C \cap ((A'\cup{1}) \times (A' \cup {1}))| \leq \frac{|C'|}{2} + a = \frac{|C'| + 2a}{2} \underset{b \geq a}\leq \frac{|C'| + a + b}{2} \leq \frac{|C|}{2}$

Y dado que $|C \cap (B' \times B')| = |C' \cap (B' \times B')| \leq \frac{|C'|}{2} \leq \frac{|C|}{2}$

Entonces tomando $A = A' \cup \{1\}, \ B = B'$ se cumple que $A \cup B = E \ \land \ A \cap B = \emptyset $, y también $\ast$. $\quad \checkmark$


Por lo tanto, por principio de inducción, para todo conjunto $E$ de estudiantes, y todo conjunto $C \subsetneq E \times E$ de pares de estudiantes, existen particiones $A, B$ que cumplen lo pedido.

$\hspace{25cm} \huge\boxtimes$

Algoritmo:

Entrada: arreglo de estudiantes $E$, arreglo de pares de estudiantes cercanos $C$

Salida: particiones $A$, $B$ tal que $A \cup B = E \ \land \ A \cap B = \emptyset \ \land \ (|C \cap (A \times A)|, |C \cap (B \times B)| \leq \frac{|C|}{2}) $

```python
#Algoritmo

def division_pandemica(E, C):
    A = [False for _ in range(len(E))]
    B = [False for _ in range(len(E))]

    pares_cercanos = {}
    for (i,j) in C:
        if i in pares_cercanos:
            pares_cercanos[i].append(j)

        elif i not in pares_cercanos:
            pares_cercanos[i] = [j]

        if j in pares_cercanos:
            pares_cercanos[j].append(i)

        elif j not in pares_cercanos:
            pares_cercanos[j] = [i]

    for i in range(len(E)):
        cantidad_cercanos_A = 0
        cantidad_cercanos_B = 0
        cercanos_a_i = pares_cercanos.get(i, [])
        for estudiante in cercanos_a_i:
            if A[estudiante]:
                cantidad_cercanos_A += 1
            if B[estudiante]:
                cantidad_cercanos_B +=1

        if cantidad_cercanos_A <= cantidad_cercanos_B:
            A[i] = True

        else:
            B[i] = True

    A = [i for i, val in enumerate(A) if val]
    B = [i for i, val in enumerate(B) if val]


    return A,B


print(division_pandemica([0,1,2,3], [(0,1), (1,2), (2,3)]))
```

#Punto 33: Maxmex

Se define la función $mex : \mathcal{P}(\mathbb{N_0}) → \mathbb{N_0}$ como

$mex(X) = mín \{j : j \in \mathbb{N_0} \ \land \ j \notin X \}$

Intuitivamente, $mex$ devuelve, dado un conjunto $X$, el menor número natural que no está en $X$. Por ejemplo, $mex(\{0,1,2 \}) = 3, mex(\{0,1,3\}) = 2 \ $ y $ \ mex(\{1,2,3,...\}) = 0.$

Dado un vector de número $a_1...a_n$ queremos encontrar la permutación $b_1...b_n$ de los mismos que maximize

$\sum_{i=1}^n mex ( \{b_1,...,b_n \} )$

Por ejemplo, si el vector es $\{3,0,1\}$ podemos ver que la mejor permutación es $\{0,1,3\}$, que alcanza un valor de $mex(\{0\}) +mex(\{0,1 \}) +mex(\{0,1,3 \}) = 1+2+2 = 5$

  Quiero escribir un algoritmo _greedy_ que resuelva el problema.

  Pensemos lo siguiente: dado un conjunto $X$ de $n$ elementos, ¿cuál es el máximo valor que puede alcanzar $mex(X)$?

  Es claro que, $mex(X) \leq n$, y se consigue que $mex(X) = n \ $ si $ \ i \in X \ \ \forall \ 0 \leq i < n$

  Es decir, $\forall X \subset \mathcal{P}(\mathbb{N_0})$, si $ \#X = n \Rightarrow mex(X) \leq n $

  De esta manera, si tengo dos conjuntos $X, Y$ tal que $X \subseteq Y$ luego se tiene que $ \# X \leq \# Y$ y, por lo tanto, $mex(X) \leq mex(Y)$

  Luego, planteo la siguiente estrategia: dado un conjunto $X$ de tamaño $n$, ordeno el conjunto de menor a mayor, y de esta manera, puedo lograr maximizar

  $\sum_{i=1}^n mex ( \{x^{(1)},...,x^{(i)} \} )$

  donde $x^{(j)}$ representa al $j$-ésimo menor elemento de $X$.

Algoritmo en pseudo-código

Entrada: arreglo $X$

Salida: arreglo permutado $X' = \{x^{(1)}, ..., x^{(n)} \}$ que maximiza  $\sum_{i=1}^n mex ( \{x^{(1)},...,x^{(i)} \} )$

$conjunto\_X$ <- $[ \ ]$

$\hspace{1cm}$ si $(i == n \ \land \ X\_ord[i] \neq X\_ord[i-1]) \ \lor \ (i \neq n \ \land \ (X\_ord[i] \neq X\_ord[i+1]))$ :

$\hspace{2cm} conjunto\_X\_ordenado$.append($X\_ord[i]$)

$\hspace{1cm}$ sino:

$n$ <- $len(X)$

saltos <- $[ \ ]$

Mientras que $i < n+1$:

$\hspace{1cm}$ si buscado == $X\_ord[i]$ :

$\hspace{2cm} conjunto\_X\_ordenado$.append($X\_ord[i]$)

$\hspace{2cm}$ buscado += $1$

$\hspace{2cm} i$ += $1$

$\hspace{1cm}$ o si $X\_ord[i] > $buscado:

$\hspace{2cm}$ saltos.extend($X\_ord[i:]$)

$conjunto\_X\_ordenado$.extend(saltos)

Devolver $conjunto\_X\_ordenado$, suma

Vamos a demostrar que este algoritmo resuelve el problema.

$Teorema:$ el algoritmo _greedy_ devuelve una solución óptima.

$Dem(\text{Greedy Stays Ahead})$:

Basta con ver que, dado $X \subset \mathcal{P}(\mathbb{N_0})$ de tamaño $n$ aribitrario, tomando $\mathcal{O} = \{o_1,...,o_n\}$ una solución óptima, y $\mathcal{G} = \{g_1,...,g_n\}$ la solución _greedy_, entonces para todo $1\leq i \leq n$ se cumple que $mex (\{o_1,...,o_i\}) \leq mex (\{g_1,...,g_i\})$

**Caso base i=1**: notar que

$\text{mex(\{x\})} = \begin{cases} 1 & x = 0 \\ 0 & x > 0
\end{cases}$

(Notar que, $\forall x \in X, \ x \geq 0$)

Luego, por definición, $g_1 = min \{x \in X \}$

Se deduce entonces que, $mex(\{g_1\}) \geq mex (\{o_1\}) \quad \checkmark$


**Hipótesis inductiva**: supongamos que $\forall 1 \leq j < n$ se cumple que $mex(\{o_1,...,o_j\}) \leq mex (\{g_1,...,g_j \})$


**Paso inductivo**: veamos que $mex (\{o_1,...,o_j,o_{j+1} \}) \leq mex(\{g_1,...,g_j, g_{j+1}\})$


Como vimos, si tengo un conjunto $A \subseteq B \Rightarrow mex(A) \leq mex(B)$

Por otra parte, si $ \# A = n$ vimos que $mex(A) \leq n$.

Entonces, en principio, si $mex (\{o_1,...,o_{j+1} \}) \leq mex (\{g_1,...,g_j \})$ no hay nada que probar, pues $mex (\{g_1,..., g_j \}) \leq mex (\{g_1,...,g_j, g_{j+1} \})$

Supongamos entonces que $mex(\{o_1,...,o_{j+1}\}) > mex (\{g_1,...,g_j\})$.

Llamemos $m = mex(\{o_1,...,o_{j+1}\})$

Luego, eso quiere decir que $\{n \in \mathbb{N_0} \mid 0 \leq n < m \} \subseteq \{o_1,...,o_{j+1}\}$

Pero, dado que $\{n \in \mathbb{N_0} \mid 0 \leq n < m \} \subseteq \mathcal{O} \Rightarrow \{n \in \mathbb{N_0} \mid 0 \leq n < m \} \subseteq X$

Y entonces, por definición del algoritmo _greedy_, $g_i = i-1 \quad \forall 1 \leq i < m$

Nótese que si $m < j+1$ entonces $m = mex(\{o_1,...,o_{j+1}\}) > mex (\{g_1,...,g_j\}) \geq mex (\{g_1,...,g_{m-1} \}) = m \quad ABS$

Y luego, si $m = j+1$ eso significa que $\{n \in \mathbb{N_0} \mid 0 \leq n \leq j \} \subseteq X$, y luego por definición del algoritmo _greedy_ ocurrirá que $g_{i} = i-1 \quad \forall 1 \leq i\leq j+1$

Y entonces $mex(\{g_1,...,g_j, g_{j+1} \}) = j+1 = max(mex(A \subset \mathcal{P}(\mathbb{N_0}) \mid \# A = j+1)) \geq mex(\{o_1,...,o_{j+1}\}) \quad \checkmark$

$ $

Por lo tanto, $mex(\{g_1,...,g_i \}) \geq max(\{o_1,...,o_i\}) \quad \forall 1 \leq i \leq n$

Y luego $\sum_{i=1}^n mex(\{g_1,...,g_i\}) \geq \sum_{i=1}^n mex(\{o_1,...,o_i\})$


Y como $n \in \mathbb{N}$ era arbitrario, entonces aplica para todo $n \in \mathbb{N}$.


$\hspace{26cm} \huge\boxtimes$

```python
# Implementación algoritmo greedy

def maxmex(X):
    n = len(X)
    elementos = [0 for _ in range(n)]
    saltos = []
    for elemento in X:
        if elemento < n and elementos[elemento] != 0:
            elementos[elemento] += 1
        elif elemento < n:
            elementos[elemento] = 1
        else:
            saltos.append(elemento)

    buscado = 0
    maximo_alcanzado = 0
    i = 0

    while buscado < n:
        if elementos[buscado] != 0:
            saltos.extend([buscado] * (elementos[buscado] - 1))
            buscado +=1
        else:
            maximo_alcanzado = buscado
            for i in range(maximo_alcanzado, n):
                saltos.extend([i] * elementos[i])
            buscado = n+1


    m = n - len(saltos)
    suma = maximo_alcanzado*(maximo_alcanzado + 1)/2 + (n - m)*maximo_alcanzado

    permutacion = [i for i in range(maximo_alcanzado)]
    permutacion.extend(saltos)

    return permutacion, suma

print(maxmex([0,0,1,1, 3, 3,4]))
```

Notar que la complejidad temporal de esta implementación es $O(n)$ dado que en principio, se recorre la lista original ($O(n)$) y luego, en el bucle _while_, si bien hay un bucle _for_ dentro, notar que, por definción, en conjunto todo tiene un costo amortiguado de $O(n)$ (y por una única vez se entra al bucle _for_).

En definitiva, como simplemente se recorre un par de veces la lista original, y se crean otro par de arreglos de dimensión $n$, entonces en cojunto tengo una complejidad temporal $O(n)$.
